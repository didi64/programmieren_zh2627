### For-Schleifen und die `range`-Funktion

Eine Schleife ermöglicht das mehrfache Ausführen eines **Codeblocks** (Schleifenbody).
In Python gibt es For- und While-Schleifen. Die While Schleife gibt
die Zahlen `0`, `1` und `2` aus, die For-Schleife sie Buchstaben `'a'`, `'b'` und `'c'`.

```python
j = 0           # Initialisierung
while j < 3:    # Doppelpunkt nach der Bedingung!
    print(j)    # Schleifenbody, eingeruckt!
    j = j + 1   # dito


for c in 'abc':  # Doppelpunkt nach der Bedingung!
    print(c)     # Schleifenbody, eingeruckt!
```

Eine While-Schleife wird wiederholt, solange eine Bedingung gilt.
(oben: solange $j <3$). Das kann (unbeabsichtigt) eine **Endlosschleife** sein!

Eine For-Schleife wir immer ein feste Anzahl Mal ausgeführt (oben 3-Mal).
For-Schleife terminieren immer. Auch braucht man sich nicht um die Initialisierung der 
Laufvariable (Schleifenvariable) zu kümmern. Deshalb sind
For-Schleifen einfacher zu handhaben. Wir betrachten vorerst nur For-Schleifen.

Eine `For`-Schleife weist der Laufvariable
bei jedem neuen Schleifendurchlauf
das **nächste Element** eines sog. **Iterable** zu.

Ein Iterable ist ein Objekt, bei dem es Sinn macht, nach einem 
nächste Element zu fragen.
Zum Beispiel ist ein String ein Iterable. 


Folgender Code führt den **eingerückten Codeblock** 
3 Mal aus. Beim 1. Durchlauf hat `character` den Wert `'a'`, dann `'b'` und dann `'c'`.
Nachdem `character` alle Zeichen des Strings durchlaufen hat, wird die Schleife verlassen. 
Nun wird der Code nach der Schleife (falls vorhanden) ausgeführt.  
Die Variable `character` enthält nun das letzte Element des Strings.

```python
for character in 'abc':
    print(character, end=', ')  # Schleifenbody

print(character)  # Wert von character nach Ende der Schleife
```

### Die `range`-Funktion

Oft will man über Zahlen von $0$ bis $n{-}1$ iterieren.
Die Funktion `range(n)` liefer ein Iterable, welches diese Zahlen **erst bei Bedarf erzeugt**.
De Aufruf `range(300_000_000)` kostet nicht mehr als `range(3)`.

```python
for i in range(3):
    print(i, end=',')
```    

Manchmal will man Code nur mehrmals ausführen und
macht keine Gebrauch von der Schleifenvariable.
In diesem Fall ist es Konvention, die Schleifenvariable mit `_` zu benennen (das ist ein gültiger Variabelname!).

```python
for _ in range(3):
    print('Test')
```    

Die Funktion `range` kann mit 1 bis 3 Argumenten aufgerufen werden:
```python
start = 2
stop = 10
step = 3

range(stop)               # 0, 1, ..., 9 (bis vor stop)
range(start, stop)        # 2, 3,..., 9  (von start bis vor stop)
range(start, stop, step)  # 2, 5, 8  (von start bis vor stop, mit Schrittweite step)
```
**Bemerkung**: Ist `stop` kleiner als `start` muss `step` negativ sein, ansonsten ist der 
erzeugte Bereich leer.

***
$n$ Fakultät berechnen ($1\cdot 2\cdot 3\cdots (n-1)\cdot n$) 

Im Notebook `L01/Erste_Programme.ipynb` haben wir ein "Programm" geschreiben, dass die
Fakultät berechnet. Dort haben wir die Zelle mit den Anweisungen
```python
result = result * i
i = i + 1
```
von Hand mehrfach ausgeführt.
Nun Verwenden wir den Zellcode als Schleifenbody.
Eine direkte Übersetzung sieht so aus:

````python
n = 5
result = 1
i = 1

for _ in range(0, n):
    result = result * i
    i = i + 1

result
````

***
Natürlich müssen wir die Schleifenvariable `i` nicht von
Hand initialisieren und inkrementieren:
***

````python
n = 5
result = 1

for i in range(1, n+1):
    result = result * i

result
````

***
Wort ohne Slicenotation in zwei Teile spalten.
***

````python
word = 'foobar'
m = 3
n = len(word)

head = ''
tail = ''

for i in range(m):
    head = head + word[i]

for j in range(i+1, n):
    tail = tail + word[j]


head, tail
````

Wort umdrehen:
***

````python
word = 'foobar'
reversed_word = ''
i = 0
n = len(word)

for i in range(n):
    character = word[i]
    reversed_word = character + reversed_word

reversed_word
````

***
Schleifen lassen sich auch schachteln:
***

````python
NEWLINE = '\n'
SEP = ' '
board = ''
cols = 'abcdefgh'

for row in range(8, 0, -1):
    board_row = ''
    for col in cols:
        field = f'{col}{row}'
        board_row = board_row + f'{field}{SEP}'

    board = board + board_row + NEWLINE

print(board)
````
