### Bedingte Anweisungen (`if-elif-else`)

Die Keywords `if`, `elif` und `else` ermöglichen es, den Ablauf eines Programms zu steuern.
Bestimmte Anweisungen oder Codeblöcke werden nur ausgeführt, falls eine bestimmte Bedingung
gilt (d.h. ein bestimmter Ausdruck) *wahr* ist.

Eine bedingte Anweisung hat folgende Form, wobei
die `elif`- und `else` Blöcken optional sind.


```python
if  <Ausdruck>: 
    <Anweisungen>
elif <Ausdruck>:   # optional 
    <Anweisungen>
⋮
elif <Ausdruck>:   # optional 
    <Anweisungen>
else:              # optional 
    <Anweisungen>
```


- Ist der Ausdruck  nach `if` *wahr* (*truthy*),
wird der nachfolgende Codeblock ausgeführt und dann ans Ende des ganzen Blocks gesprungen. 

- Ist der Ausdruck *false* (*falsy*)), wird zum nächsten `elif`-Block gesprungen und wieder gleich verfahren.
Gibt es keinen `elif`-Block, wird der Code im `else`-Block ausgeführt.


**Beispiele**
1. Vorzeichen einer Zahl berechnen:
```python
x = ...

if x > 0:
    sign_x = 1
elif x < 0:
    sign_x = -1
else:
    sign_x = 0

sign_x
```

2. [Fizzbuzz](https://en.wikipedia.org/wiki/Fizz_buzz).
Bei diesem Trinkspiel wird reihum gezählt, $1, 2, \ldots$
Statt der Zahl $n$ muss man manchmal ein bestimmtes Wort sagen:
- *fizz*, falls $n$ durch 3 teilbar aber nicht durch 5,
- *buzz*, falls $n$ durch 5 teilbar aber nicht durch 3, 
- *fizzbuzz*, falls $n$ durch 3 und 5 teilbar.

```python
n = 13
answer = ''

if n % 3 == 0 and n % 5 == 0:
    answer = 'fizzbuzz'
elif n % 3 == 0:
    answer = 'fizz'
elif n % 5 == 0:
    answer = 'buzz'
else:
    answer = str(n)
```

````python
N = 20
answers = ()

for n in range(1, N):
    if n % 3 == 0 and n % 5 == 0:
        answer = 'fizzbuzz'
    elif n % 3 == 0:
        answer = 'fizz'
    elif n % 5 == 0:
        answer = 'buzz'
    else:
        answer = str(n)

    answers = answers + (answer,)

answers[:7]
````
