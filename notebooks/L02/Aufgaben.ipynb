5. Teste, ob eine Zahl $n$ eine Primzahl ist.

   Range $(2, \lceil\sqrt{n}\rceil)$ hat.

````python
number = 899
is_prime = True
bound = int(number**0.5 + 1)

for i in range(2, bound):
    if number % i == 0:
        is_prime = False
        break

is_prime, i
````

````python
word = 'abakadabra'
character = 'a'

occurrences = ()
n = len(word)

for i in range(n):
    c = word[i]
    if c == character:
        occurrences = occurrences + (i,)

occurrences
````

1. Eine Geheimzahl soll erraten werden. 
Der Benutzer wird solange nach einer Zahl gefragt, bis er die Geheimzahl erraten hat.
Nach jedem Versuch wird mitgeteilt, ob er zu klein, zu gross oder richtig geraten wurde. 

````python
lower_bound = 0       # inklusiv
upper_bound = 10_000  # exklusiv
secret = 4750

for i in range(upper_bound):
    guess = (lower_bound + upper_bound) // 2

    if guess == secret:
        break
    if guess < secret:
        lower_bound = guess+1
    elif guess > secret:
        upper_bound = guess

    if lower_bound == upper_bound:
        break

print('Anzahl Versuche:', i+1, 'Zahl:', guess)
````

````python
lower_bound = 0       # inklusiv
upper_bound = 10_000  # exklusiv
secret = 4750

while lower_bound < upper_bound:
    guess = (lower_bound + upper_bound) // 2

    if guess == secret:
        break
    if guess < secret:
        lower_bound = guess+1
    elif guess > secret:
        upper_bound = guess

    if lower_bound == upper_bound:
        break

print('Anzahl Versuche:', i+1, 'Zahl:', guess)
````

````python
lower_bound = 0       # inklusiv
upper_bound = 10_000  # exklusiv
secret = 4750

for i in range(upper_bound):
    guess = (lower_bound + upper_bound) // 2

    if guess == secret:
        break

    is_too_small = guess < secret
    is_too_big = not is_too_small

    lower_bound = is_too_big*lower_bound + is_too_small*(guess + 1)
    upper_bound = is_too_small*upper_bound + is_too_big*guess

print('Anzahl Versuche:', i+1, 'Zahl:', guess)
````
