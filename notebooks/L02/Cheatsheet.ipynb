### Cheatsheet

### Typen
- **int** und **float**: Integer/ ganze Zahlen '42' und Floats/Fliesskommazahlen `1.5`.  
Operationen: `+, -, *, /, ** (hoch), // (Ganzzahldivision), % (modulo)`  
Funktionen: `abs(x), min(x, y), max(x, y), round(x)`.  
**str**: String/Zeichenkette `"Ein String"` oder `'Ein String'`  
Operationen: `'foo' + 'bar'`, `50 * '-'`.  
Funktionen: `len(s)`.  
<!-- **f-Strings** `f'Zahl {x}, Quadrat: {x*x}'`  
Der Ausdrücke in `{...}` werden ausgewertet und in einen String umgewandelt. -->

- **bool**: `True` und `False` (verhalten sich wie `1` und `0`)
`x` ist *wahr* oder *truthy*, falls `bool(x)` gleich `True`, sonst *falsch* oder *falsy*.   
Nur `0`, `0.0`, `None` und leere Iterables (`''`, `()`, `[]`, `{}`, ...) sind *falsch*.  
Operationen: `and`, `or`, `not`, `+`, `*`, `-`.  
- **tuple**: Tuple `x = ('Max', 'Muster')`.  
Operationen: `+`, `*` (Multiplikation mit Integer).  
Funktionen: `len`.  

**Typ eines Objekts**: `type(x)` liefert den Typ von `x`. Jeden Typ gibt es nur einmal, deshalb werden Typen mit `is` verglichen.
```python
type(x) is int  # True
```

**Casts**: (Typumwandlung) Ist `tp` ein Typ, liefert `tp(obj)` nach Möglichkeit ein Objekt vom Typ `tp`.  
```python
str(42)       # '42'
tuple('abc')  # ('a', 'b', 'c')
```

**Index und Slicenotation**:  
Verwendung: `s[stop], s[start:stop], s[start:stop:step]` fehlt start und/oder stop wird
vom Anfang bez. bis zum Ende alles herausgegriffen.
- `s[0],  s[-1], s[:2]`


**`range`**:  
Verwendung: `range(stop), range(start, stop)` oder `range(start, stop, step)`

**For-Loops**:  
```python
for number in range(5):
    print(number)
```
- `continue`: nächsten Schleifendurchlauf starten,
- `break`: Schleife verlassen.
