### Binäre Suche

Folgendes Suchproblem taucht oft auf.  
**Problem**: Sei `names` ein **sortiertes** Tuple mit Namen.
Kommt der Name `'Diana'` darin vor?

Enthält das Tuple nur wenige Namen,
kann man mit einem For-Loop über alle Namen im Tuple iterieren.
```python
target = 'Diana'

for name in names:
    if name == target:
        break

name == target  # True, falls target im Tuple
```

Dies ist sehr ineffizient, wenn das Tuple sehr viele Namen enthält, denn man muss
im Worst-Case alle Elemente des Tuples anschauen.
Da das Tuple sortiert ist, gibt es eine effizientere Suchstrategie:    
Wir versuchen, den Index `idx` mit `names[idx] == Diana` zu erraten.

Sei `n` die Länge des Tuples, und `lower_bound = 0` (inklusiv), `upper_bound = n` (exklusiv).

Solange `lower_bound < upper_bound`,  
raten wir den Index in der Mitte, `idx = (lower_bound + upper_bound) // 2`.

- Ist `names[idx]` zu klein, haben wir zu klein geraten.  
  Wir setzen `lower_bound = idx + 1`.
- Ist `names[idx]` zu gross, haben wir zu gross geraten.  
  Wir setzen `upper_bound = idx`. 


Suchen wir einen Index im Range (0, 7), sieht das so aus:
```
      3         # 1. Versuch, nach links, falls zu klein, sonst rechts
    /   \ 
  1       5     # 2. Versuch, dito
 / \     / \
0   2   4   6   # 3. Versuch, dito

```

Zuerst testen wir den Suchalgorithmus gründlich an einem
Tuple mit wenigen Elementen und geben zusätzliche Infos aus,
die beim Fehersuchen (Debuggen) helfen.

````python
names = ('a', 'c', 'e', 'g', 'i', 'k', 'm')

lower_bound = 0             # inklusiv
upper_bound = len(names)    # exklusiv
target = 'c'


for i in range(upper_bound):
    print('Attempt:', i+1, 'searching in range', (lower_bound, upper_bound))
    idx = (lower_bound + upper_bound) // 2
    name = names[idx]
    print(name, 'at index', idx)

    if name == target or lower_bound >= upper_bound:
        break
    if name < target:          # zu klein
        lower_bound = idx+1    # inklusiv
    elif name > target:        # zu gross
        upper_bound = idx      # exklusiv

    if lower_bound == upper_bound:
        break


success = names[idx] == target


print('\nlooked up', i+1, 'names')
if success:
    print(target, 'is at index', idx)
else:
    print(target, 'not found!')
````
