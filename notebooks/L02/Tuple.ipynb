### Tuple
Ähnlich wie ein String, der eine Folge von Zeichen speichert,
speichert ein Tuple eine Folge von Werten. Die Werte dürfen
von unterschiedlichem Typ sein.
Wie Strings sind Tuple **immutable**.
Einzelne Elemente eines Tuples kann man nicht modifizieren!

- Tuples lassen sich wie Strings mit `+` aneinanderhängen und mit `*` vervielfachen.  
- `len` liefert die Länge (Anzahl Elemente) des Tuples
- Index und Slicenotation funktioniert genau wie bei Strings.
-  Zwei Tuple sind gleich bez. `==` falls ihre entsprechenden Elemente bez. `==` gleich sind.  
Und wie bei Strings entscheidet das 1. Element, bei dem sich die Tupel unterscheiden, welches das kleinere Tuple ist.
- Ein String `s` lässt sich mit `tuple(s)` in ein Tuple von Zeichen umwandeln. 

```python
names = ('Alice', 'Bob') 
prime_numbers = (2, 3, 5, 7, 11, 13)
person = ('Max', 'Muster', 1995)
letters = tuple('abc')

# Spezialfälle:
empty_tuple = ()                  # leeres Tuple
einelementiges_tuple = ('Bob',)   # Das Komma ist obligatorsich!
    
names[0]
prime_numbers[:3]
```



**Bemerkung**:  
Die äusseren Klammern sind beim Erzeugen eines Tuples (oft) fakultativ.


**Beispiel**:  
Positionen der Vorkommen in Tuple speichern.
***

````python
# Initialisierung
word = 'abca'
character = 'a'

i = 0
n = len(word)
occurrences = ()
````

````python
occurrences, n
````

````python
# Zelle n mal ausfueheren
# Erstellt Tuple mit Positionen, wo character in word vorkommt
c = word[i]
is_match = c == character
occurrences = occurrences + (i,) * is_match  # Tuple (i,) anhaengen.
i = i + 1
````
