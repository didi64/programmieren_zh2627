### Weitere Typen (`bool` und `tuple`), Vergleichsoperatoren
Integers, Floats, Strings und Funktionen sind alles sog. Objekte.
Jedes Python Objekt `obj` hat einen Typ. `type(obj)` liefert diesen Typ. 
```python
type(5)
```
Manche Objekte lassen sich in ein Objekt eines anderen Typs umwandeln.
Das Umwandeln bezeichnet man als *casting* oder *cast*. 
Den String `'42'` kann man wie folgt in einen `int` umwandeln:
```python
int('42')  # 42 (Integer)
# str(42)  # '42' (String)
```
Ist die Umwandlung nicht möglich, wird ein Fehler ausgelöst.

### Wahrheitswerte
Die speziellen Werte `True` und `False` sind vom Typ `bool` (Boolean/Wahrheitswert).
`True` und `False` verhalten sich in jeder Hinsicht wie `1` und `0`.
Wahrheitswerte sind in diesem Sinne spezielle Integers.
Man verwende jedoch `True` und `False` um 
klar zu machen, dass man nun `1` und `0` als Wahrheitswerte interpretiert.
```python
type(True)             # bool
isinstance(True, int)  # True
isinstance(1, bool)    # False
```
Die Funktion `isinstance(<obj>, <type>)` liefert `True`, falls sich
das Objekt `<obj>` in jeder Beziehung wie ein Objekt vom Typ `<type>`
verhält, sonst `False`.

**Operationen** mit Wahrheitswerten:  Wahrheitswerte lassen mit `not` negieren und mit 
`and` und `or` verknüpfen:
```python
x = True
y = False

not x     # bool(1 - x)
x and y   # bool(x*y)
x or y    # bool(x + y)
```

**Wichtig**: Jedes Objekt `obj` kann mit `bool(obj)` in einen Wahrheitswert verwandelt werden!  
Es gilt: `bool(obj)` ist `True`, ausser `obj` ist
>  `0`, `0.0`,  `None` oder ein Iterable ohne Elemente, also  
>  `''` (leerer String), `()` leeres Tuple, ...

Ein Ausdruck `expr` nennen wir *truthy* oder *wahr*, falls `bool(expr)` `True` ist, sonst *falsy* oder *falsch*.  
```python
bool('')  # False
bool(0)   # False
bool('0') # True!
```

### Vergleichsoperatoren
Die Vergleichsoperatoren 

> `==` (gleich), `!=` ($\not =$), `<`, `>`, `<=`, `>=` (für Zahlen, Strings, ...)
> `is` und `is not` (für Typen)

versuchen 2 Objekte zu vergleichen und liefern im Erfolgsfall `True` oder `False`,
andernfalls wird ein Fehler ausgelöst.

**Wichtig**: Man muss unterscheiden: 
`=` ist der Zuweisungsoperator, `==` ist ein Vergleichsoperator.

```python
2 != 3                # False
'Anfang' < 'Ende'     # True
type(2) is int        # True
type(2) is not str    # False
```

### Tuple
Ähnlich wie Strings, die eine Folge von Zeichen speichern,
kann ein Tuple eine Folge von Werten speichern. Die Werte können dabei 
einen unterschiedlichen Typ haben.
Wie Strings sind Tuple **immutable**.
Einzelene Elemente eines Tuples kann man nicht modifizieren.

- Tuples lassen sich wie Strings mit `+` aneinanderhängen und mit `*` vervielfachen.  
- `len` liefert die Länge (Anzahl Elemente) des Tuples
- Index und Slicenotation funktioniert genau wie bei Strings.
-  Zwei Tuple sind gleich bez. `==` falls ihre entsprechenden Elemente bez. `==` gleich sind.  
Und wie bei Strings entscheidet das 1. Element, bei dem sich die Tupel unterscheiden, welches das kleinere Tuple ist.
- Ein String `s` lässt sich mit `tuple(s)` in ein Tuple von Zeichen umwandeln. 

```python
names = ('Alice', 'Bob') 
prime_numbers = (2, 3, 5, 7, 11, 13)
person = ('Max', 'Muster', 1995)
letters = tuple('abc')

# Spezialfälle:
empty_tuple = ()                  # leeres Tuple
einelementiges_tuple = ('Bob',)   # Das Komma ist obligatorsich!
    
names[0]
prime_numbers[:3]
```



**Bemerkung**:  
Die äusseren Klammern sind beim Erzeugen eines Tuples fakultativ.  
Wir haben das beim AUsgeben mehrere Werte schon benutzt!

```python
1, 2 + 3  # ist eigentlich das Tuple (1, 2 + 3)
```

Das ist manchmal bequem, kann aber zu Fehlern führen und ist nur selten guter Stil.


**Aufgaben**:

````python
word = 'abca'
character = 'a'

i = 0
n = len(word)
count = 0
````
