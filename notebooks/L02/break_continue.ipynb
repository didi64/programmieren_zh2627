### `break` und `continue`



- Mit `break` kann sofort aus einer For-Schleife ausgebrochen werden.
- Mit `continue` kann sofort der nächste Schleifendurchlauf gestartet werden.
  
**Bemerkung**: Dies gilt auch für **While**-Schleifen!

***
Ein typisches Problem ist das Suchen
eine Elementes in einem String oder Tuple.
Dabei iteriert man mit einem For-Loop über die Elemente
und testet mit `==` ob es sich um das gesuchte Element handelt.
Abhängig vom Resultat verlässt man die Schleife oder setzt die Sucht fort.
***

```python
from data import names

len(names), names[:3], names[-3:]
```

```python
target = 'Benjamin'

for name in names:
    if name == target:
        print('found!')
        break

if name != target:
    print('not found')
```

```python
word = 'abcdefgh'
character = 'd'

result = False
for c in word:
    if c == character:
        result = True
        break

result
```

```python
word = 'abcdefgh'
character = 'x'

result = False
i = 0
for c in word:
    if c == character:
        result = True
        break
    i = i + 1

result, i
```

```python
word = 'abakadabra'
character = 'a'

result = ''
for c in word:
    if c == character:
        continue
    result = result + c

result
```
