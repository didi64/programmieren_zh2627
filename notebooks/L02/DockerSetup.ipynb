### Docker Container, Security-Issue

Der aktuelle Container Setup erlaubt es jedem,
der sich im gleiche lokalen Netz befindet, 
auf den Jupyterlab-Server zuzugreifen.

Um dies zu unterbinden, ist folgendes zu tun:


🛠️ Schritt 1: Erstmalige Einrichtung (Nur einmal ausführen)

Diesen Schritt musst du nur beim ersten Mal ausführen, um die Umgebung zu erstellen.
1. Öffne die Windows-Eingabeaufforderung (Klicke auf Start, tippe cmd ein und drücke Enter).
2. Navigiere in deinen spezifischen Kursordner, in dem du deine Notebooks speichern möchtest.
   Bist du in deinem Home Verzeichnis `C:\Users\YourUsername` oder ähnlich, 
    > `cd programmieren`
    > 
    sonst
    > `cd /d %USERPROFILE%`  
    > `cd programmieren`

  
3. Vergewissere dich, dass du dich im richtigen Ordner befindest, bevor du den nächsten Schritt ausführst, da sich Docker genau auf dieses Verzeichnis festlegt!  
Kopiere und füge den folgenden Befehl exakt so ein, wie er unten steht, und drücke Enter:
   ```cmd
   docker run --name jupyter -d -p 127.0.0.1:10000:8888 --hostname jlab -v .:/home/studi/work dieterprobst/jlab:3.2
   ```
   
Deine Umgebung ist nun erstellt und läuft unbemerkt im Hintergrund!


🚀 Schritt 2: Auf JupyterLab zugreifen

Sobald der Container läuft, kannst du über deinen Webbrowser eine Verbindung herstellen.
1. Öffne deinen Browser und gehe auf: `http://localhost:10000`


🔄 Schritt 3: Täglicher Gebrauch (Nächstes Mal starten)

Du musst den langen Einrichtungsbefehl nie wieder eingeben. Wenn du deinen PC herunterfährst oder Docker schliesst, bleibt dein Container gespeichert.

1. Öffne die Anwendung Docker Desktop.
2. Klicke auf den Reiter Containers (falls nötig)
3. Suche `jupyter` in der Liste und klicke rechts auf das blaue Start (Play)-Symbol.
4. Rufe `http://localhost:10000` in deinem Browser auf oder klicke auf das [↗] Icon.
