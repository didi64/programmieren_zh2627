📘 JupyterLab Docker Einrichtungsanleitung

Diese Anleitung zeigt dir, wie du deine JupyterLab-Umgebung für diesen Kurs einrichtest. Nach der Konfiguration werden deine Arbeiten lokal auf deinem Computer gespeichert und deine Umgebung ist vollständig vor unbefugtem Zugriff aus dem Netzwerk geschützt.

🛠️ Schritt 1: Erstmalige Einrichtung (Nur einmal ausführen)

Diesen Schritt musst du nur beim ersten Mal ausführen, um die Umgebung zu erstellen.
1. Öffne die Windows-Eingabeaufforderung (Klicke auf Start, tippe cmd ein und drücke Enter).
2. Navigiere in deinen spezifischen Kursordner, in dem du deine Notebooks speichern möchtest. Zum Beispiel:cmd
cd C:\Benutzer\DeinName\Dokumente\work
Use code with caution.
(Vergewissere dich, dass du dich im richtigen Ordner befindest, bevor du den nächsten Schritt ausführst, da sich Docker genau auf dieses Verzeichnis festlegt!)
3. Kopiere und füge den folgenden Befehl exakt so ein, wie er unten steht, und drücke Enter:cmd
docker run --name docki -d -p 127.0.0.1:10000:8888 --hostname jlab -v .:/home/studi/work <dockerimage>
Use code with caution.
(Hinweis: Ersetze <dockerimage> durch den genauen Image-Namen, den deine Lehrperson bereitgestellt hat).
Deine Umgebung ist nun erstellt und läuft unbemerkt im Hintergrund!

🚀 Schritt 2: Auf JupyterLab zugreifen

Sobald der Container läuft, kannst du über deinen Webbrowser eine Verbindung herstellen.
1. Öffne deinen Browser und gehe auf: http://localhost:10000
2. Den Login-Token finden: Wenn JupyterLab beim ersten Mal nach einem Passwort oder Token fragt:
	• Öffne das Programm Docker Desktop.
	• Klicke links auf den Reiter Containers.
	• Klicke auf den Container mit dem Namen docki.
	• Schaue in den Reiter Logs. Dort siehst du weiter unten Zeilen, die so ähnlich aussehen: http://127.0.0...
	• Kopiere die lange Kette aus Buchstaben und Zahlen hinter token= und füge sie im Browser in das Login-Feld ein.

🔄 Schritt 3: Täglicher Gebrauch (Nächstes Mal starten)

Du musst den langen Einrichtungsbefehl nie wieder eingeben. Wenn du deinen PC herunterfährst oder Docker schliesst, bleibt dein Container gespeichert.

Methode A: Über die Docker Desktop App (Am einfachsten)

1. Öffne die Anwendung Docker Desktop.
2. Klicke auf den Reiter Containers.
3. Suche docki in der Liste und klicke rechts auf das blaue Start (Play)-Symbol.
4. Rufe http://localhost:10000 in deinem Browser auf.

Methode B: Über die Eingabeaufforderung (Am schnellsten)

1. Öffne die Eingabeaufforderung (es ist egal, in welchem Ordner du dich dieses Mal befindest).
2. Führe diesen kurzen Befehl aus:cmd
docker start docki
Use code with caution.
3. Rufe http://localhost:10000 in deinem Browser auf.

⚠️ Fehlerbehebung & Zurücksetzen

Falls du den Befehl zur Ersteinrichtung versehentlich im falschen Ordner ausgeführt hast oder deine Umgebung komplett zurücksetzen möchtest, kannst du diesen Befehl in der Eingabeaufforderung ausführen, um den alten Container zu löschen:
cmd
docker rm -f docki
Use code with caution.
Gehe danach zurück zu Schritt 1, stelle sicher, dass du im richtigen work-Ordner stehst, und führe den langen docker run-Befehl erneut aus.
