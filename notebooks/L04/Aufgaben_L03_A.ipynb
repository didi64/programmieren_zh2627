1. Zähle mit Hilfe einer For-Schleife
die Anzahl Vorkommen eines Zeichens in einem String (ohne Verwendung bedingter Anweisungen).
```python
word = 'abca'
character = 'a'
count = 0

2
```

````python
# Vergleichsop, Vorkommen eines Zeichens in
# einem String zaehlen
word = 'abca'
character = 'a'
count = 0

n = len(word)
for i in range(n):
    is_match = (word[i] == character)
    count = count + is_match

count
````

4. Das Tuple `fizzbuzz` enthält die ersten 20 Antworten
   des Spiels Fizzbuzz. Erstelle selber ein Tuple mit den
   Antworten, und teste, ob dein Tuple die gleichen Werten enthält.
   Hier die Regeln:
    - `'fizz'`, falls $n$ durch 3 teilbar aber nicht durch 5,
    - `'buzz'`, falls $n$ durch 5 teilbar aber nicht durch 3, 
    - `'fizzbuzz'`, falls $n$ durch 3 und 5 teilbar.
    - die aktuelle Zahl als String, sonst. 
```python
fizzbuzz = (
    '1', '2', 'fizz', '4', 'buzz',
    'fizz', '7', '8', 'fizz', 'buzz',
    '11', 'fizz', '13', '14',
    'fizzbuzz', '16', '17', 'fizz', '19',
)

N = 20
answers = ()
```

````python
fizzbuzz = (
    '1', '2', 'fizz', '4', 'buzz',
    'fizz', '7', '8', 'fizz', 'buzz',
    '11', 'fizz', '13', '14',
    'fizzbuzz', '16', '17', 'fizz', '19',
)

N = 20
answers = ()

for n in range(1, N):
    if n % 3 == 0 and n % 5 == 0:
        answer = 'fizzbuzz'
    elif n % 3 == 0:
        answer = 'fizz'
    elif n % 5 == 0:
        answer = 'buzz'
    else:
        answer = str(n)

    answers = answers + (answer,)

answers == fizzbuzz
````

8. Gib nun das letzte Tier im String `animals` aus.
   Das sollte auch dann funktionieren, wenn `animals` nur ein Tier
   enthält (suche nach dem letzen Vorkommen von Space in `animals`,
   teste ob `animals[-1]` oder `animals[-2]` ... gleich Space ist).
```python
SPACE = ' '
animals = 'Ape Bee Cat Dog Eagle'

'Eagle'
```

````python
SPACE = ' '
animals = 'Ape Bee Cat Dog Eagle'

n = len(animals)
for i in range(1, n+1):
    if animals[-i] == SPACE:
        break

if animals[-i] == SPACE:
    last_animal = animals[-i+1:]
else:
    last_animal = animals

print('last animal:', last_animal)
````

11. Kombiniere den Code der beiden obigen Aufgaben, um alle Spaces vor und nach einem  Wort zu entfernen.
```python
SPACE = ' '
word = '   foobar  '

'foobar'
```   

````python
SPACE = ' '
word = '   foobar  '

n = len(word)

for i in range(n):
    if word[i] != SPACE:
        break

for j in range(1, n+1):
    if word[-j] != SPACE:
        break

word_strippt = word[i:n-j+1]
word_strippt
````

12. Ersetze in einem String jedes Komma durch ein Semicolon.
```python
OLD_SEP = ','
NEW_SEP = ';'

words = 'alpha, beta, gamma, delta'

'alpha; beta; gamma; delta'
```

````python
OLD_SEP = ','
NEW_SEP = ';'

words = 'alpha, beta, gamma, delta'
words_new = ''

n = len(words)
for i in range(n):
    if words[i] != OLD_SEP:
        words_new = words_new + words[i]
    else:
        words_new = words_new + NEW_SEP

words_new
````
