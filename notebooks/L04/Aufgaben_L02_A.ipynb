6. Spalte ein Wort in zwei Teile `head` und `tail` ohne Slicenotation zu nutzen.

````python
# 6
word = 'foobar'
m = 3

head = ''
tail = ''
n = len(word)

for i in range(m):
    head = head + word[i]

for j in range(i+1, n):
    tail = tail + word[j]

head, tail, head == word[:m], tail == word[m:]
````

9. Speichere in einem Tupel, an welchen Positionen ein Buchstabe in einem Wort vorkommt.
   Teste dann mit einem For-Loop, ob
   an diesen Positionen tatsächlich dieser Buchstabe steht.

````python
# 9
word = 'abakadabra'
character = 'a'

occurrences = ()
n = len(word)

for i in range(n):
    c = word[i]
    if c == character:
        occurrences = occurrences + (i,)

occurrences
````

````python
result = True

for i in occurrences:
    if word[i] != character:
        result = False
        break

result
````

10. Teste, ob eine Zahl $n$ eine Primzahl ist.
   Teste mit einem For-Loop, ob die Zahl einen Teiler im
   Range $(2, \lceil\sqrt{n}\rceil)$ hat.

````python
# 10
number = 899
is_prime = True
bound = int(number**0.5 + 1)

for i in range(2, bound):
    if number % i == 0:
        is_prime = False
        break

is_prime, i
````

12. Unten importieren wir das Tuple names aus `data.py`.
    Benutze den Code aus dem Notebook `Binaere_Suche.ipynb`, um im Tuple `names` einen Namen zu finden.

````python
from data import names


type(names), len(names)
````

````python
# 11
lower_bound = 0             # inklusiv
upper_bound = len(names)    # exklusiv
target = 'David'

for i in range(upper_bound):
    idx = (lower_bound + upper_bound) // 2
    name = names[idx]

    if name == target or lower_bound >= upper_bound:
        break
    if name < target:          # zu klein
        lower_bound = idx+1    # inklusiv
    elif name > target:        # zu gross
        upper_bound = idx      # exklusiv

    if lower_bound == upper_bound:
        break


success = names[idx] == target


print('\nlooked up', i+1, 'names')
if success:
    print(target, 'is at index', idx)
else:
    print(target, 'not found!')
````
