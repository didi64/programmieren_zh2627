### Funktionen
Damit eine Programmiersprache von praktischem Nutzen ist, muss es möglich sein,
bestehenden und getesteten Programmcode einfach wiederverwenden zu können. 
**Funktionen** ermöglichen genau das.

Eine Funktionsdefinition weist einem Codeblock (Funktionsbody) einen Namen zu und gibt an, welche Variabeln beim Funktionsaufruf initialisiert werden sollen. 
Im Funktionsbody können zudem **return**-Anweisungen stehen. 
Der Wert des Ausdrucks nach dem Keyword `return` wird dann
beim Aufruf der Funktion zurückgegeben. 
Steht kein Wert nach `return`, dann wird `None` zurückgegeben. Wird das Ende des Funktionsbodys erreicht, wird ebenfalls der Wert `None` zurückgegeben (ohne explizites `return`).




**Funktionsdefinition**  
```python
def add(x, y):
    '''gibt die Summe der Zahlen x und y zurueck'''
    return x + y
```

**Funktionsaufruf**
```python
add(2, 3)          # 5   (2 + 3)
add(x=2*3, y=5+5)  # 16  (6 + 10)
```

Beim Funktionsaufruf `add(2, 3)` werden den (lokalen) Variabeln `x` und `y` der Funktion `add(x, y)` die
Werte 2 und 3 zugewiesen. 
Der Funktionsbody wird dann mit diesen Werten ausgeführt. 
Der Wert des Ausdrucks `add(2, 3)` ist der Wert des mit `return`
zurückgegebenen Ausdrucks `x + y` (5).


**Zusatz**:
Funktionsargumenten kann man Default-Werte zuweisen.
Die Default-Argumente müssen nach den normalen (*positional arguments*) stehen.  
Wird beim Funktionsaufruf einem Default-Argument **kein**
Wert übergeben, wird der Default-Wert verwendet, sonst der übergebene Wert.


```python
def find(s, sub, start=0, end=None):
    '''gib den ersten Index von Substring sub 
       in s[start:end] zurueck oder -1
    '''
```
Der Triple-quoted String (**Doc-String**) direkt nach der Funktionssignatur ist optional.
Er wird angezeigt, falls auf den die Funktion geklickt und `shift-tab` gedrückt wird.  

Der Funktionsbody dar nicht leer sein.
```python
def f():  # fehlender Body erzeugt Fehler!

def f():
    pass  # ok, pass ist eine Anweisung, die nichts tut


def f():
    '''to do'''  # ok, Doc-String ist Teil des Bodys

    
def f():
    ...  # (Ellipses, spezieller Wert) ok
```

### Funktionen haben ihre eigenen (lokalen) Variabeln
Wird im Funktionsbody eine Variable  
`<Variabelname> = <Ausdruck>`  

definiert, dann ist `<Variabelname>` eine **lokale Variable**
(auch wenn `<Variabelname>` bereits eine globale Variable ist).

- Lokale Variablen sind nur innerhalb der Funktion sichtbar/modifizierbar.
- Eine Funktion kann globale Variabeln lesen, aber nicht überschreiben.

```python
x = 2  # globale Variabeln
y = 3  # dito

def g(z):                      # z ist  eine lokale Variable  
    y = 43                     # y wird eine lokale Variable
    print(f'globales x={x}')   # Zugriff auf globales x
    print(f'lokales y={y}')
    print(f'lokales z={z}')

    x = 23                     # Fehler: x ist keine lokale Variable
                               # globale Variable kann nicht geschrieben werden
```

````python
def say_hello(name):
    print(f'hello, {name}!')
````

````python
say_hello('Bob')
````

````python
s = say_hello('Max')
print(s)
````

````python
def is_palindrom(word):
    '''word: str
       gibt True zurueck, falls das Wort ein Palindrom ist
    '''
    return word == word[::-1]
````

````python
is_palindrom('sugus')
````
