### Lambda-Funktionen (von [$\lambda$-Calculus](https://en.wikipedia.org/wiki/Lambda_calculus))  
Oft beschreibt man eine einfache Funktionen wie folgt:
$$
x \mapsto x+1 \text{ (Nachfolger)},\qquad 
y \mapsto y^2 \text{ (Quadrat)},\quad 
(x, y)\mapsto x + y \text{ (Summe)}.
$$
Das oder die Funktionsargumente stehen links von Pfeil.
Beim Anwenden der Funktion auf eine Zahl/Zahlen (die Funktionsargumente),
werden diese im Ausdruck rechts vom Pfeil eingesetzt. Dann wird der
Wert des Ausdrucks berechnet.  
Wohl weil kompakter und einfacher zum Drucken hat sich
folgende Variante etabliert.

$$
\lambda\, x{:}\,x+1 \text{ \,\,(Nachfolger)},\qquad 
\lambda\, y{:}\,y^2 \text{  \,\,(Quadrat)},\quad 
\lambda\, x, y{:}\, x + y \text{   \,\,(Summe)},
$$

welche als sogenannte Lambda-Ausdrücke bez. Lambda-Funktionen 
auch vom Python-Interpreter verstanden werden.
```python
successor = lambda x: x + 1
square = lambda y: y**2
add = lambda x, y: x + y
```

Die Variable `successor` enthält nun die Nachfolgerfunktion.
Um den Nachfolger oder die Summe zu berechnen, muss man die Funktion
mit den entsprechenden Argumenten **aufrufen**. 
```python
number = 2

successor(number)    # number + 1
square(number)       # number**2 
add(2, 3)            # 2 + 3 
```

Ein Lambda-Ausdruck wird kompakt so beschreiben:
```python
lambda <Argumente>: <Ausdruck>
```
Dabei sind `<Argumente>` und  `<Ausdruck>` durch
konkrete Argumente und einen konkreten Ausdruck zu ersetzen.
`<Argumente>` ist dabei eine möglicherweise leere, komma-separierte Liste
von Variabelnamen.

**Zusatz**:
Per Default werden Argumente nach Position übergeben (**positional arguments**):
Das erste Argument im Funktionsaufruf wird dem
ersten Argument/Variable nach dem `lambda` in der Funktionsdefinition zugewiesen, u.s.w.  
Argumente können auch als sog. **Keyword-Paare** übergeben werden
(müssen nach den *positional arguments* stehen):
```python
add(2, 3)      # ok, x=2, y=3
add(2, y=3)    # ok, dito
add(x=2, y=3)  # ok, dito
add(x=2, 3)    # SyntaxError: positional argument follows keyword argument
```

### Beispiele zum Testen
**Betrag**:
Wie weit ist es von $10$ nach $7$ auf der $x$-Achse?  
Die Distanz ist $\operatorname{abs}(10 - 7) = \operatorname{abs}(7 - 10) = 3$.
```python
get_distance1 = lambda x, y: abs(y - x)
```

**Distanz vom Punkt $p=(x_1, y_1)$ zum Punkt $q=(x_2, y_2)$**:
```python
get_distance2 = lambda p, q: ((q[0] - p[0])**2 + (q[1] - p[1])**2)**0.5
```

**Vorzeichen** berechnen:  
Das Vorzeichen (*sign*) einer Zahl ist $1$ (Zahl positiv),
$-1$ (Zahl negativ) oder $0$.
```python
sign = lambda x: (x > 0) - (x < 0)
```

**Substring in String einfügen**:
```python
insert_word = lambda s, sub, i: s[:i] + sub + s[i:]
```

**Zahl in bestimmten Bereich halten**:  
- `cap(z)` reduziert $z$ auf den Maximalwert 100, falls $z$ grösser ist.
- `lift(z)` erhöht $z$ auf den Minimalwert 50, falls $z$ kleiner ist.
- `clamp(z)` reduziert oder erhöht, falls ausserhalb [50,..., 100].
```python
cap = lambda z: min(z, 100)
lift = lambda z: max(z, 50)
clamp = lambda z: cap(lift(z))  # lambda-Funktionen kann man schachteln
```
**Schranken mit Konstanten festlegen**:
```python
LOWER_BOUND = 20
UPPER_BOUND = 40

cap = lambda z: min(z, UPPER_BOUND)
lift = lambda z: max(z, LOWER_BOUND)
clamp = lambda z: cap(lift(z))
```
