### Vorgehen zum Schreiben einer Funktion
Code ist schnell geschrieben, die Fehlersuche dagegen ist oft sehr zeitintensiv.
Deshalb versucht man, den **Funktionsbody** aus bereits **getesteten** Codefragmenten zusammenzusetzen.  

Exemplarisch erstellen wir eine Funktion, welche ein Wort umdreht. 
Um die Funktion bei Bedarf verwenden zu können, speichern wir diese dann in einem `.py`-File `my_functions.py` im Ordner `modules`.  
Dabei stützen wir uns auf folgenden Code.

````python
word = 'abcd'

reversed_word = ''
for c in word:
    reversed_word = c + reversed_word

reversed_word
````

***
Mit obigem Code können wir einen String umdrehen:
Wir müssen den String einfach
anstelle von `'abcd'` der Variable `word` zuweisen und dann die Zelle ausführen. 
Der umgedrehte String ist dann in der Variable `reversed_word` gespeichert.    

Um eine Funktion zu erhalten, welche diese Aufgabe löst, 
definieren wir eine Funktion, die `word` als Argument nimmt und
im Funktionsbody `reversed_word` berechnet und zurück gibt.  

Die Funktion können  wir ebenfalls `reversed_word` nennen.
Das verursacht kein Problem, da der Funktionsname eine globale Variable und 
der Rückgabewert eine lokale Variable ist.

```python
def reverse_word(word):
    reversed_word = ''
    for c in word:
        reversed_word = c + reversed_word
    
    return reversed_word
```

Dann fügen wir noch den DocString ein und testen die Funktion mit einigen Wörtern.
***

````python
def reverse_word(word):
    '''word: str -> str
       gibt das umgedrehte Wort zurueck
    '''
    reversed_word = ''
    n = len(word)
    for i in range(n):
        reversed_word = word[i] + reversed_word

    return reversed_word
````

````python
# Funktion testen
word = 'Sugus'
reverse_word(word) == word[::-1]
````

**Funktion in `.py`-File speichern**  
Um die Funktion bei Bedarf verwenden zu können, speichern wir diese 
in einem `.py`-File (z.B. `string_helpers.py`) im Ordner `modules`.  
(Navigiere in den Ordner `modules`, öffne/erstelle das File `string_helpers.py` und
copy/paste die Funktion in dieses File).   

Dann können wir diese Funktion in einem unserer Notebooks mit
> `string_helpers.reverse_word('hallo')`

aufrufen, falls vorgängig
```python
import string_helpers
```

erfolgreich ausgeführt wurde. Verursacht obige Anweisung einen Fehler, so ist der Fehler im File
