###  Module, das Modul `random`

In Python kann man bestehenden und getesteten Programmcode einfach wiederverwenden.  
Dazu speichert man den Code in einem `.py` File. So ein File nennt man auch **Modul**. Der **Modulname** ist der Filename ohne die `.py` Endung.  

**Modul als Ganzes importieren**:  
```python
import <Modulname>
```
Dieses Kommando importiert das Modul `<Modulname>`. Dabei geschieht Folgendes:  
- Das File `<Modulname>.py` wird ausgeführt. Dabei wird ein eigener **Namesraum** verwendet. Alle Variabeln und Funktionen, die
definiert werden, sind mit `<Modulname>.<Variabelname>` anzusprechen.

**Einzelne Namen/Objekte aus Modul importieren**:
```python
from data import names
```
Dieses Kommando importiert die aufgelisteten Namen.
Obige Anweisung hat den gleichen Effekt wie

```python
import data
names  = data.names
```

Das Modul wird **nur beim ersten Import** ausgeführt.
Soll das Modul ein weiteres Mal ausgeführt werden, z.B. weil es modifiziert wurde, muss der **Kernel neu gestartet werden**.


Typischerweise speichert man in einem Modul Programmcode, welchen man wiederverwenden will.
In einem File `string_helpers.py` könnte man z.B. Funktionen zum 
Arbeiten mit Strings speichern.

Die **Python Standard Library** besteht aus einer Vielzahl von Modulen, wie z.B. das Modul `random`, welches
Funktionen zum Arbeiten mit zufälligen Werten enthält. 
Nach dem Import des Moduls stehen diese Funktionen zur Verfügung.

```python
import random


# 3 Zufallszahlen von 1 bis und mit 10 ausgeben
for i in range(3):
    number = random.randint(1, 10)
    print(f'{i+1}. Zufallszahl: {number}')


# 4-stelliger Pin generieren
pin = ''
for _ in range(4):
    digit = random.randint(1, 9)
    pin = pin + str(digit)

pin


# Tuple mit 10 Zufallsziffern erstellen
digits = ()
for _ in range(10):
    digit = random.randint(1, 9)
    digits = digits + (digit,)

digits
```

### Wo wird nach Modulen gesucht?
Die Variable `path` des Moduls `sys` ist eine Liste mit Verzeichnissen, in denen 
nach dem zu importierenden Modul gesucht wird.
Der leere String `''` steht für den das aktuelle Arbeitsverzeichnis.

Die Suche beginnt beim ersten Verzeichnis. Das erste Modul, das gefunden wird, wird importiert.
Wird das Modul nicht gefunden wird ein `ModuleNotFoundError` erzeugt.
