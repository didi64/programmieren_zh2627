### Listen
Listen sind **veränderbare** (mutable) Tuples.  
Eine Liste wird wie folgt erzeugt:
```python
empty_list = []                # leere Liste
items = [1, 2, 'foo', (1, 2)]  # Liste
```
Wie bei Tuples,  liefert 
```python
len(items)   # die Länge (Anzahl Elemente) der Liste
items[i]     # das i-te Element
items[:i]    # die ersten i Elemente
```
Im Gegensatz zu Tuple lassen sich die Elemente von
Listen modifizieren:
```python
items[i] = 42 # weist items[i] neuen Wert zu
```
- Listen lassen sich mit `+` aneinanderhängen und
mit `*` vervielfachen.
Dabei wird eine **neue** Liste erstellt.


- Zwei Listen sind gleich bez. `==` falls ihre entsprechenden Elemente bez. `==` gleich sind.
Wie Tupels lassen sich Listen mit `<` und `<=` vergleichen.
