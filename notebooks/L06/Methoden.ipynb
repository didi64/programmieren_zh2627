### Methoden

In Python sind Integer, String, Tuple, Listen, Funktionen, ...  sogenannte Objekte eines bestimmten Typs. `type(obj)` liefert den Typ des Objekts `obj`.

Jeder Typ stellt Funktionen zur Verfügung, mit welchen sich die Objekte
dieses Typs manipulieren lassen. Ähnlich wie bei einem Modul
lassen sich diese Funktionen mit der sogenannten Dot-Notation ansprechen.
Der Typ `str` stellt u.a. folgende Funktionen bereit:

```python
str.upper('foo')       # 'FOO'
str.find('abcde', 'b') # 1
str.strip('  foo  ')   # 'foo'
```

Tippt man `str.` und dann `tab` in einer Codezelle, werden u.a. die
Funktionen des Tys `str` angezeigt und man kann eine Funktion auswählen.
<img src='../../images/stringmethoden.png'>
<br>
Klickt man dann `shift-tab` wird der Hilfstext zur Funktion angezeigt. 
Alternativ kann man die Funktion `help` benutzen:
```python
help(str.upper)  # Hilfstext zu str.upper anzeigen
```

**Methoden**  
Ist `s` ein String und `f` eine Funktion des Typs `str`, 
dann ist `s.f(...)` die Kurzform von
`str.f(s, ...)`.  **Beachte**: `s` wird als zusätzliches erstes Argument
eingefügt! 
Wird `f` auf diese Weise aufgerufen, spricht man von einem Methodenaufruf.
Man sagt auch *`f` ist eine String-Methode*, falls
man `f(s, ...)` via `s.f(...)` aufrufen kann.

```python
s = 'abceda'
s.upper()       # str.upper(s)
s.find('a')     # str.find(s, 'a')
s.find('a')     # str.find(s, 'a', )
s.find('a', 1)  # str.find(s, 'a', 1)
```

**Methodchaining**: Die Methode `upper` gibt ein Objekt vom Typ `str` zurück, auf das mit der **dot**-Notation direkt eine weitere Methode angewendet werden kann. 
```python
'foo'.upper().center(10)
'abcd'.replace('a', 'X').replace('b', 'Y')
```
