### Guessing Game
1. Implementiere eine Funktion
   `get_msg(guess, secret, n_try)`, welche  einen String zurück gibt.
   Der String soll mitteilen, ob der Rateversuch zu gross, zu klein oder korrekt war.
   Im letzteren Fall soll so auch noch die Anzahl Versuche mitgeteilt werden.
2. Generiere eine Zufallszahl. Frage dann mit einer For-Schleife 1000 mal nach einer Zahl. Gib nach jedem Rateversuch mit Hilfe obiger Funktion ein Feedback.
   Verlasse die Schleife, falls die Zahl erraten wurde.
3. Schreibe eine Funktion `guessing_game(lower=1, upper=100)`, die
    eine Zufallszahl im Range (lower, upper+1) wählt, die dann der Benutzer erraten muss. Benutze dann statt der For-Schleife eine While-Schleife.
   Löse das Problem mit Ausbrechen aus der While-Schleife, einer Hilfsvariable und einer direkten Bedingung.

```python
def get_msg(guess, secret, n_try):
    if guess == secret:
        msg = f'Congrats, you guessed {secret} in {n_try} tries.'
    elif guess < secret:
        msg = f'{guess} is too small'
    else:
        msg = f'{guess} is too big'

    return msg
```

```python
get_msg(guess=5, secret=5, n_try=2)
```

```python
import random

random.randint(1, 100)
```

```python
lower = 1
upper = 100
secret = random.randint(lower, upper)
guess = None


for i in range(1000):
    guess = int(input('Zahl?'))
    msg = get_msg(guess, secret, i+1)
    print(msg)

    if guess == secret:
        break
```

```python
def guessing_game(lower=1, upper=100):

    secret = random.randint(lower, upper)
    guess = None

    for i in range(1000):
        guess = int(input('Zahl?'))
        msg = get_msg(guess, secret, i+1)
        print(msg)

        if guess == secret:
            break
```

```python
def guessing_game_1(lower=1, upper=100):
    secret = random.randint(lower, upper)
    count = 0

    while True:
        guess = int(input('Zahl?'))
        count = count + 1
        msg = get_msg(guess, secret, count)
        print(msg)

        if guess == secret:
            break
```

```python
def guessing_game_2(lower=1, upper=100):
    secret = random.randint(lower, upper)
    count = 0
    done = False

    while not done:
        guess = int(input('Zahl?'))
        count = count + 1
        msg = get_msg(guess, secret, count)
        print(msg)

        done = guess == secret
```

```python
def guessing_game_3(lower=1, upper=100):
    secret = random.randint(lower, upper)
    guess = None
    count = 0

    while guess != secret:
        guess = int(input('Zahl?'))
        count = count + 1
        msg = get_msg(guess, secret, count)
        print(msg)
```
