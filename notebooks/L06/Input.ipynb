### Die Funktion `input` 

Die Funktion `input(prompt=''`) erlaubt es, den Benutzer zu Tastatureingaben aufzufordern und diese entgegenzunehmen. Die Eingabe wird als String zurückgegeben.

```python
answer = input('Eingabe:')
answer  #  eingebenener Text
```

### Der Hilfstext zu `input`
```
Signature: input(prompt='')  
(der Hilfstext sagt  input(prompt='', /), aber korret ist input(prompt=''))  
Docstring:
Read a string from standard input.  The trailing newline is stripped.

The prompt string, if given, is printed to standard output without a
trailing newline before reading input.
```


Die Signatur  `input(prompt='')`  zeigt an, dass die Funktion ein Argument mit Defaultwert `''` hat (leerer String).
Dieses Argument kann überschrieben werden.  
Die Funktion `input` kann 3 Arten aufgerufen werden:

```python
input()
input('Name?')
input(prompt='Name?')
```

**Wichtig**: input() fordert den Benutzer zur Eingabe von Text auf. Das Ausführen weitere Zellen ist bockiert, bis die Eingabe erfolgt ist. Wird versucht eine Zelle auszuführen während auf eine Eingabe gewertet wird, wird folgende Mitteilung gemacht:

>Cell not executed due to pending input
The cell has not been executed to avoid kernel deadlock as there is another pending input! Submit your pending input and try again.


### Aufgaben
Schreibe Code, der
1. den Benutzer zweimal nach einer Zahl fragt und dann die Summe ausgibt.
1. den Benutzer nach Name und Jahrgang fragt, und dann ausgibt:  
   `Hallo ..., du bist ... Jahre alt.`

````python
x = int(input('1. Zahl'))
y = int(input('2. Zahl'))

print(f'{x} + {y} ist {x+y}')
````

````python
name = input('Name:')
jahrgang = int(input('Jahrgang:'))

print(f'Hallo {name}, du bist {2026-jahrgang} Jahre alt.')
````
