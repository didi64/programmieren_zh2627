### Bubble-Sort
Bubble-Sort ist ein **einfacher**, aber **ineffizienter** Algorithmus zum Sortieren von Listen.
Eine Liste $[a_1, a_2, \ldots, a_n]$ heisst (aufsteigend) sortiert,
falls $a_1\leq a_2\leq ...\leq a_n$.
***
**Bubble-Sort Algorithmus**  

Gehe von links nach rechts durch die Liste.
Falls das aktuelle Element grösser ist als sein rechter Nachbar,
vertausche die Elemente.
Dies bringt das grösste Element auf seine richtige Position (ans Ende der Liste).  
```python
[4,3,2,1] -> [3,4,2,1] -> [3,2,4,1] -> [3,2,1,4]
```

Auf die gleiche Weise sortieren wir nun noch die ersten $n-1$ Elemente.
```python
[3,2,1,4] -> [2,3,1,4] -> [2,1,3,4]  # bringe 3 auf seinen Platz
[2,1,3,4] -> [1,2,3,4]               # bringe 2 auf seinen Platz
                                     # 1 bereits auf seinem Platz
```
***
Wieviele Vergleiche sind nötig
um eine Liste mit $n$ Elementen zu sortieren?
$$(n-1)+(n-2)+\cdots +1=\frac{n^2-n}{2}\simeq \frac{n^2}{2}.$$

<br>
Allgemein gilt: $1+2+\cdots+(n-1)=\frac{n^2-n}{2}$.

### Aufgabe 2
Nutze nun, dass `bubble_up(items)`
das grösste Element ans Ende der Liste schiebt.
Eine weitere Anwendung von `bubble_up(items)` braucht das
letzte Element der Liste nicht mehr zu besuchen.


```python
def bubble_sort(items):
    n = len(items) - 1
    for i in range(n):
        bubble_up(items, k=i)


def bubble_up(items, k=0):
    '''Gehe von links nach rechts items.
       Ignoriere die letzten k Elemente von items!
       Falls items[i] > items[i+1], vertausche die Elemente.

       Bsp.
       xs = [4,3,2,1]
       bubble_up(xs, k=1)
       xs  # [3,2,4,1]
    '''
```

### Divide and Conquor
Divide and Conquor ist eine Lösungstrategy bei der das Problem auf ein oder mehrere
einfachere Teilprobleme des gleichen Typs reduziert wird.

Beim Bubble-Sort sieht dies so aus.
Sortiere eine Liste mit $n$ Elementen wie folgt:

Ist $n\leq 1$ ist nichts zu tun.
Anderfalls, schiebe das grösste Element ganz nach rechts.  
Es bleibt noch das einfachere Problem: Sortiere die
ersten n-1 Elemente der Liste.
   
**Rekursive Variante** von Bubble-Sort:  
Eine Funktion kann auch sich selber aufrufen (Rekursion).  


```python
def recursive_bubble_sort(items, k=0):
    '''sortiere items, ignoriere die letzten k Elemente'''
    if len(items) - k < 2:  # Liste sortiert
        return
        
    bubble_up(items, k)  # schiebe max. relevantes El. nach rechts
    recursive_bubble_sort(items, k=k+1)  # loese Restproblem
```
