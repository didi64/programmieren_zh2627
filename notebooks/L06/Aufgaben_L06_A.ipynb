2. Erstelle eine Liste mit den Strings `['1-Mal', ... ,'9-Mal']`.
   Gib die ersten und letzten 3 Elemente aus.  
   Schreibe dann eine Funktion `get_header()`, die den
   String
   > `'1-Mal 2-Mal 3-Mal 4-Mal 5-Mal 6-Mal 7-Mal 8-Mal 9-Mal'`
   
   zurück gibt.

````python
items = []
for i in range(1, 10):
    items.append(f'{i}-Mal')

items[3:], items[-3:]
````

````python
def get_header():
    items = []
    for i in range(1, 10):
        items.append(f'{i}-Mal')

    return ' '.join(items)
````

3. Erstelle eine Liste mit den Vielfachen von 3.  
   `[3, 6, 9, 12, 15, 18, 21, 24, 27]`.
   Erstellen eine Liste mit den Vielfachen von 3 als Strings der
   Länge 6 (benutze `str.ljust`).  
   `['3     ', ... '27    ']`.  
   Schreibe dann eine Funktion `get_multiples(n)`, welche zuerst eine Liste wie oben mit
   den Vielfachen von `n` als Strings der
   Länge 6 produziert, dann diese Strings mit `join` verbindet und das Resultat 
zurück gibt.

```python
get_multiples(n)  # '3     6     9     12    15    18    21    24    27    '
```

````python
numbers = []
n = 3

for i in range(1, 10):
    # numbers.append(i*n)
    numbers.append(str(i*n).ljust(6))

numbers[:2], numbers[-2:]
````

````python
def get_multiples(n):
    numbers = []
    for i in range(1, 10):
        numbers.append(str(i*n).ljust(6))

    return ''.join(numbers)
````
