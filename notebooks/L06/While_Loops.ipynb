### While-Schleifen
Eine While-Schleife ist eine Anweisung der Form  
```python
while <Ausdruck>:
    <Anweisungen>
```
Die `<Anweisungen>` im im eingerückten Codeblock werden wiederholt, solange der Wahrheitswert von `<Ausdruck>` wahr ist. Mit der Anweisung `break` im Schleifenbody kann die Schleife vorzeitig verlassen werden.

**Bemerkungen**: While-Schleifen können (oft unbeabsichtigt) **Endlosschleifen** sein!  
Mit Interrupt Kernel im Kernel Menu ist die Zelle zu stoppen.  

While-Schleifen treten in folgenden Varianten auf:
- nach `while` steht explizit eine Bedingung,
- nach `while` steht eine Hilfsvariable, mit dem Wert `True`, die dann später einmal im `While`-Body auf `False` gesetzt wird, sobald die Schleife nicht mehr durchlaufen werden soll.
- man startet die Schleife mit `while True` und bricht dann mit `break` aus.


````python
# direkte Bedingung

counter = 3
while counter >= 0:
    print(counter, end=', ')
    counter = counter - 1
````

````python
# Hilfsvariable

counter = 3
done = False

while not done:
    print(counter, end=', ')
    counter = counter - 1
    if counter < 0:
        done = True
````

````python
# Ausbrechen

counter = 3
while True:
    print(counter, end=', ')
    counter = counter - 1
    if counter < 0:
        break
````

````python
PIN = '1234'
pin = input('Pin:')

while PIN != pin:
    print('falscher Pin.')
    pin = input('Pin:')

print('Login was successful!')
````
