```python
def get_bounds(names):
    min_len = 100
    max_len = 0

    for name in names:
        min_len = min(min_len, len(name))
        max_len = max(max_len, len(name))

    return (min_len, max_len)


def get_n_letter_names(names, n):
    results = ()
    for name in names:
        if len(name) == n:
            results = results + (name,)

    return results


def get_groups(names):
    bounds = get_bounds(names)
    groups = ()
    for i in range(bounds[0], bounds[1]+1):
        group = get_n_letter_names(names, i)
        groups = groups + (group,)

    return groups
```

```python
import random


def get_random_pin():
    pin = ''
    for _ in range(4):
        digit = str(random.randint(1, 9))
        pin = pin + digit

    return pin


def get_random_numbers(n, lower=1, upper=9):
    numbers = ()
    for _ in range(n):
        number = random.randint(lower, upper)
        numbers = numbers + (number,)

    return numbers
```

```python
def reverse_word(word):
    reversed_word = ''
    for c in word:
        reversed_word = c + reversed_word

    return reversed_word
```

```python
words = ('foobar', 'Sugus', 'abcd')
failures = 0

for word in words:
    if word[::-1] != reverse_word(word):
        print('Failure for word', word)
        failures = failures + 1


n = len(words)
msg = str(n-failures) + '/' + str(n) + ' Tests passed'
print(msg)
```

```python
def delete_character(word, character):
    '''word: str
       character: str der Laenge 1
       entfernt aus word alle Vorkommmen von character
       und gibt das resulierende Wort zurueck
    '''
    result = ''
    for c in word:
        if c == character:
            continue
        result = result + c

    return result
```

```python
def get_positions(word, idxs):
    '''word: str
       idxs: tuple with indices
       returns the string formed of
       the characters word[i] for i in idxs

       get_positions('abcde', (2, 4)) -> 'ce'
    '''
    result = ''
    for i in idxs:
        result = result + word[i]

    return result
```

```python
def strip(word):
    '''entfernt allen Space vor und nach dem Wort
       und gibt dann das Wort zurueck
    '''
    SPACE = ' '
    n = len(word)

    for i in range(n):
        if word[i] != SPACE:
            break

    for j in range(1, n+1):
        if word[-j] != SPACE:
            break

    return word[i:n-j+1]
```

```python
def get_occurrences(word, char):
    '''returns a tuple with all the positions
       where char occurs in word
    '''
    occurrences = ()
    n = len(word)

    for i in range(n):
        if (word[i] == char):
            occurrences = occurrences + (i,)

    return occurrences
```

```python
def split(s, sep=','):
    items = ()
    occurrences = get_occurrences(s, sep)

    start = 0
    for i in occurrences:
        item = s[start:i]
        item = strip(item)
        items = items + (item,)
        start = i+1

    return items
```

```python
def find(s, sub, start=0, end=None):
    '''gib den ersten Index von Substring sub
       in s[start:end] zurueck oder -1
    '''
    s = s[start:end]
    if s == '' or sub == '':
        return -1              # early return

    n = len(s)
    m = len(sub)
    end = end or n
    for i in range(n):
        if s[i:i+m] == sub:
            return start + i   # index, falls gefunden

    return -1                  # Wert, falls nicht gefunden
```

```python
def is_prime(n):
    '''gibt (True, n) zurueck, falls n eine Primzahl
       sonst (False, Teiler)
    '''
    bound = int(n**0.5 + 1)

    for i in range(2, bound):
        if n % i == 0:
            return False, i

    return True, n
```

```python
for i in range(100):
    is_prime_, divisor = is_prime(i)
    if is_prime_:
        print(f'{i} is prime')
    else:
        print(f'{i} is divisable by {divisor}')
```

```python
import random


def guessing_game(lower=1, upper=99):
    '''waehlt eine Zufallszahl in (lower, upper+1) und
       fragt solange nach einer Zahl, bis die Zahl erraten wurde.
       Gibt zurueck, ob zu gross oder zu klein geraten wurde.
    '''
    secret = random.randint(lower, upper)
    guess = int(input('Zahl?'))
    count = 1

    while guess != secret:
        if guess > secret:
            print('too big')
        else:
            print('too small')

        guess = int(input('Zahl?'))
        count = count + 1

    print(f'Congrats, you guessed the secret number {secret} in {count} tries.')
```
