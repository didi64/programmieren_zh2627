### Recap Lambda-Funktionen und Funktionen
Ein Lambda-Ausdruck (Lambda-Funktion) hat die Form
```python
lambda <Argumente>: <Ausdruck>
```
Dabei sind `<Argumente>` und  `<Ausdruck>` durch
konkrete Argumente und einen konkreten Ausdruck zu ersetzen.
`<Argumente>` ist dabei eine möglicherweise leere, komma-separierte Liste
von Variabelnamen.

````python
nachfolger = lambda n: n+1
nachfolger(3)
````

````python
add = lambda x, y: x + y
add('foo', 'bar')
````

````python
add = lambda x, y=1: x + y  # y ist Defaultargument
add(5)                      # 5 + 1
````

````python
add(5, 5)  # 5 + 5
````

````python
f = lambda: print('foo')  # Funktion mit side-effect
result = f()              # gibt None zurueck,
````

````python
print(result)  # Rueckgabewert
````

````python
identity = lambda x: x  # gibt Argument wieder zurueck
identity('test')
````

````python
insert = lambda s, sub, i: s[:i] + sub + s[i+1:]
insert('abcd', ' xxx ', 2)
````

### Funktionen
Der Code nachstehender Zelle findet einen kleinsten Wert
des Tuples `values`.  
Um eine Funktion zu definieren, welche den
kleinsten Wert eines Tupels findet,
deklarieren wir eine Funktion die values als Argument nimmt,
deren Funktionsbody das Minimum findet und einer Variable `minimum` zuweist, und
diese dann mit `return minimum` zurück gibt.
```python
def minimum(values):
    <Code der minimum berechnet>
    return minimum
```

````python
values = (5, 2, 3)

minimum = values[0]        # 1. Kandidat fuer Minimu
for value in values[1:]:   # iteriere ueber verbleibende Werte
    if value < minimum:    # aktuallisiere minimum, falls noetig
        minimum = value

minimum                    # teste Code
````

````python
def minimum_0(values):
    minimum = values[0]
    for value in values[1:]:
        if value < minimum:
            minimum = value

    return minimum
````

````python
numbers = (4, 6, 2, 1, 5)
minimum(numbers)
````

````python
values = ('x', 'aaa', 'bb')
minimum_0(values)  # liefert den alphabetisch kleinsten Wert
````

***
Nachstehend modifizieren wir unsere Funktion so, dass
ein kürzester String aus einem Tuple mit Strings
zurückgegeben wird.
Anstelle des Strings `word` betrachten wird seine 
Länge `len(word)` (Sortierwert).
***

````python
def minimum_1(words):
    '''gib ein kuerzestes Wort aus word zurueck'''
    minimum = words[0]
    for word in words[1:]:
        if len(word) < len(minimum):
            minimum = word

    return minimum
````

````python
values = ('x', 'aaa', 'bb')
minimum_1(values)
````

**Typische Verwendung von Lambda-Funktionen**    
Wir möchten nun eine
Minimumsfunktion, bei welcher der Benutzer
spezifizieren kann, was minimiert werden soll:
Die Funktion nimmt ein zusätzliches Argument `key`.
`key` ist dabei selber eine Funktion, welche jedem Wert in `values`
einen Sortierwert zuweist (in obigem Beispiel war dies die Länge).
Zurückgegeben wird ein Wert mit minimalem Sortierwert.

````python
def minimum(values, key=lambda x: x):
    '''gibt einen value aus values zurueck,
       fuer den key(value) minimal ist
    '''
    minimum = values[0]
    for value in values[1:]:
        if key(value) < key(minimum):
            minimum = value

    return minimum
````

````python
values = ('x', 'aaa', 'bb')
minimum(values)  # alphabetisch kleinster Wert
````

````python
minimum(values, key=lambda x: len(x))  # kuerzerster Wert
````

````python
minimum(values, key=len)  # kuerzerster Wert
````

````python
points = ((3, -1), (3, 1), (2, 2))
minimum(points, key=lambda p: p[0]**2 + p[1]**2)  # kleinster Abstand von Ursprung
````
