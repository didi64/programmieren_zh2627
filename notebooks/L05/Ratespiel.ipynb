### Ratespiel

Wir möchten auf folgende Weise ein Ratespiel spielen können:
1. Zelle ausführen, welche eine Zufallszahl der Variable `secret`  zuweist.
2. Zelle ausführen, welche der Variable `guess` eine Zahl zuweist und
dann ausgibt, ob zu gross, zu klein oder richtig geraten wurde.
Diese Zelle führen wir solange aus, bis wir die Zahl erraten haben.


Um eine Zufallszahl zu generiert, benutzen wir die
Funktion `randint(lower, upper)` aus dem Modul `random`.

```python
import random

help(random.randint)  # Zeigt Hilfstext zu random.randint
```

````python
import random  # Zelle, die secret Zufallszahl zuweist


lower = 1
upper = 10
secret = random.randint(lower, upper)
````

````python
guess = 5  # Zelle, die den Rateversuch guess evaluiert

if guess < secret:
    print('too small')
elif guess > secret:
    print('too big')
else:
    print('correct')
````

````python
import random


def new_game(lower, upper):
    secret = random.randint(lower, upper)
    return secret


def evaluate_guess(guess):
    if guess < secret:
        print('too small')
    elif guess > secret:
        print('too big')
    else:
        print('correct')
````

````python
secret = new_game(1, 5)  # Zelle, die secret Zufallszahl zuweist
````

````python
guess = 5                # Zelle, die guess evaluiert
evaluate_guess(3)
````
