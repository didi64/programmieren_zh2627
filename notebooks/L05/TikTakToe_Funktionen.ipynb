### TikTakToe Hilfsfunktionen
- `show_position(position)` gibt die Spielposition aus.
- `replace_char(s, char, idx)` ersetzt in `s` das Zeichen an Position `idx` durch `char` und wird zur Aktualisierung der Spielposition benutzt.
- `is_tiktak(position, player)` liefert `True`, falls `player` (X oder O) gewonnen hat. Für jede Gewinnmöglichkeit `idxs` in `winners = ((0, 1, 2), (3, 4, 5), ...)`
  wird getestet, ob an den Positionen `idxs` jeweils
  das Symbol `player` steht. Die Funktion `is_tiktaktoe` nutzt die
   Hilfsfunktionen `get_chars_at_idxs` und `do_idxs_match_char`.

4. Implementation einer Funktion
```python
def do_idxs_match_char(word, idxs, char):
    '''liefert True, falls in word an allen 
       Positionen idxs das Zeichen char steht
    '''
```
- Benutze  `get_chars_at_idxs` um die Zeichen an den Positionen `idxs` aus
  einem String `s` herauszugreifen.
  Teste, dass `s` gleich `char * len(idxs)` ist.
- Mache daraus eine Funktion.
***

````python
s = '0x2xx5'
char = 'x'
idxs = (1, 3, 4)

chars = get_chars_at_idxs(s, idxs)
chars == char * len(idxs)
````

````python
def do_idxs_match_char(word, idxs, char):
    chars = get_chars_at_idxs(word, idxs)
    # print(chars)
    return chars == char * len(idxs)
````

````python
do_idxs_match_char(s, idxs, char)
````

5. Implementation einer Funktion
```python
def is_tiktak(position, player):
    ''' liefert True, falls player gewonnen hat'''
```
- Erstelle ein Tuple `winners = ((0, 1, 2), (3, 4, 5), ...)` mit
  allen Gewinnmöglichkeiten.
  Teste mit einem For-Loop und `do_idxs_match_char`, ob ein Tuple `idxs` in
  `winners` nur das Symbol `player` enthält.
- Mache daraus eine Funktion.
***

````python
position = '...XXX...'
char = 'X'


winners = ((0, 1, 2), (3, 4, 5), (6, 7, 8))
for idxs in winners:
    if do_idxs_match_char(position, idxs, char):
        print(f'{char} wins')
````

````python
def is_tiktak(position, player):
    winners = ((0, 1, 2), (3, 4, 5), (6, 7, 8),
               (0, 3, 6), (1, 4, 7), (2, 5, 8),
               (0, 4, 8), (2, 4, 6),
               )

    for idxs in winners:
        if do_idxs_match_char(position, idxs, player):
            return True
````

````python
s = 'O...O...O'
is_tiktak(s, 'O')
````
