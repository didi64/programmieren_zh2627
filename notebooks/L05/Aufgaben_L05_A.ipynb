````python
import string_helpers


def get_chars_at_idxs(word, idxs):
    '''gibt einen String zurueck, bestehend aus
       den Zeichen word[i], fuer i in idxs

       get_positions('abcde', (2, 4)) -> 'ce'
    '''
    result = ''
    for i in idxs:
        result = result + word[i]

    return result


def get_occurrences(word, char):
    '''gibt Tuple zurueck, mit allen Positionen
       wo char in word vorkommt

       get_ocurrences('aabbab', 'a') -> (0, 1, 4)
    '''
    occurrences = ()
    n = len(word)

    for i in range(n):
        if (word[i] == char):
            occurrences = occurrences + (i,)

    return occurrences


def split(s, sep=','):
    '''Trennt den String s bei sep und
       gibt ein Tuple mit den Teilen zurueck

       split('foo, bar, baz') -> ('foo', 'bar', 'baz')
       split('foo, bar, baz,') -> ('foo', 'bar', 'baz', '')
    '''
    items = ()
    occurrences = get_occurrences(s, sep) + (len(s),)

    start = 0
    for i in occurrences:
        item = s[start:i]
        item = string_helpers.strip(item)
        items = items + (item,)
        start = i+1

    return items
````

4. Gegeben ist ein Tuple mit Feldern eines Schachbretts.
   Ein Feld ist ein String der Form 'a1' (0-te Spalte, 0-te Zeile).
   Schreibe eine Funktion `king_moves(field)`, welche die
   Anzahl Züge berechnet, die ein König braucht,
          um vom Feld 'a1' nach `field` zu gelangen.  
   Benutze dazu die Hilfsfunktion
    ```python
    def get_row(char):
        '''get_row('a') -> 1
           get_row('b') -> 2, ...
        '''
        letters = 'abcdefgh'
        for i in range(8):
            if char == letters[i]:
                return i
    ```
    Benutze dann die Funktion `maximum`, um das Feld zu finden, welches 
    am weitesten vom Feld 'a1' entfernt ist. 
   
    ```python    
    fields = ('a2', 'b4', 'c7', 'f5', 'g6')
    maximum(fields, key=king_moves)  # 'c7'
    ```   

````python
def maximum_0(values):
    maximum = values[0]
    for value in values[1:]:
        if value > maximum:
            maximum = value

    return maximum
````

````python
maximum_0((4, 10, 3))
````

````python
def maximum(values, key):
    maximum = values[0]
    for value in values[1:]:
        if key(value) > key(maximum):
            maximum = value

    return maximum
````

````python
maximum((4, -10, 3), key=abs)
````

````python
def get_row(char):
    letters = 'abcdefgh'
    for i in range(8):
        if char == letters[i]:
            return i


def king_moves(field):
    col = get_row(field[0])
    row = int(field[1])
    n_moves = max(col, row)
    return n_moves
````

````python
fields = ('a2', 'b4', 'c7', 'f5', 'g6')
maximum(fields, key=king_moves)
````
