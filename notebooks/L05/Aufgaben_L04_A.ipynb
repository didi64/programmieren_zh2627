1. Implementiere folgende Funktionen als $\lambda$-Ausdrücke:
    - `sign(x)`: $1$ (falls $x >0$), $-1$ (falls $x < 0$), $0$ sonst.
    - `get_distance(p, q)`: Distanz vom Punkt $p=(x_1, y_1)$ nach $q=(x_2, y_2)$.
    - `insert_word(text, word, i)`: Füge `word` an Position `i` in `text` ein.

```python
sign(-2)  # -1

p = (1, 2)
q = (5, 5)
get_distance(p, q)  # 5.0

text = 'das Test'
word = 'ist ein '
i = 4
insert_word(text, word, i)  # 'das ist ein Test'
```

````python
sign = lambda x: (x > 0) - (x < 0)
get_distance = lambda p, q: ((q[0] - p[0])**2 + (q[1] - p[1])**2)**0.5
insert_word = lambda s, word, i: s[:i] + word + s[i:]
````

````python
x = 2
sign(x)
````

````python
p = (1, 2)
q = (5, 5)
get_distance(p, q)
````

````python
insert_word('das Test', 'ist ein ', 4)
````

2. Eine Geheimzahl `secret` soll erraten werden. Implementiere folgende Funktion
   als $\lambda$-Ausdruck:

    - `get_feedback(guess)`: liefert den passenden String aus `feedbacks`.  
   Benutze oben definierte Funktion `sign`, um den Index des Feedbacks zu berechen. 
   
```python
feedbacks = ('correct', 'too small', 'too big')
secret = 42

guess = 50

```

````python
feedbacks = ('correct', 'too small', 'too big')
secret = 42

get_feedback = lambda guess: feedbacks[sign(secret - guess)]
````

```python
def reverse_word(word):
    '''word: str -> str
       gibt das umgedrehte Wort zurueck

```

Teste mit einigen Strings, dass jeweils `reverse_word(word) == word[::-1]`.  
Wieviele Primzahlen kleiner als 100 gibt es?

````python
def reverse_word(word):
    '''word: str -> str
       gibt das umgedrehte Wort zurueck
    '''
    reversed_word = ''
    for c in word:
        reversed_word = c + reversed_word

    return reversed_word


def is_prime(n):
    '''gibt True zurueck, falls n eine Primzahl
       sonst False
    '''
    if type(n) is not int or n < 2:
        return False

    bound = int(n**0.5 + 1)
    for i in range(2, bound):
        if n % i == 0:
            return False

    return True
````

````python
words = ('foobar', 'Sugus', 'abcd')

for word in words:
    print(reverse_word(word) == word[::-1])
````

````python
primes = ()
for i in range(100):
    if is_prime(i):
        primes = primes + (i,)

primes[:3], primes[-5:], len(primes)
````

6. Benutze die Funktion `randint` aus dem Modul `random`, 
um einen zufälligen Pin zu erstellen (z.B. `'3867'`). Implementiere dann eine Funktion
`get_random_numbers(n, lower=1, upper=9)`, welche ein
Tuple mit n zufälligen Ziffern zurückgibt (z.B. `(3, 8, 6, 7)`).

```python
import random

def get_random_pin():
    '''liefert einen zufälligen 4-stelligen Pin (z.B. '3867')'''
    
def get_random_numbers(n, lower=1, upper=9):
    '''liefert ein Tuple mit n Zufallszahlen 
       im Bereich (lower, ..., upper).

       get_random_numbers(3) -> (2, 4, 9)
       '''
```

````python
import random


def get_random_pin():
    '''lieferr einen zufälligen 4-stelligen Pin (z.B. '3867')'''
    pin = ''
    for _ in range(4):
        digit = str(random.randint(1, 9))
        pin = pin + digit

    return pin


def get_random_numbers(n, lower=1, upper=9):
    '''liefert ein Tuple mit n Zufallszahlen 
       im Bereich (lower, ..., upper).
    '''
    numbers = ()
    for _ in range(n):
        number = random.randint(lower, upper)
        numbers = numbers + (number,)

    return numbers
````

Speichere sie dann in `modules/string_helpers.py`.

```python
def delete_character(word, character):
    '''word: str
       character: str der Laenge 1
       entfernt aus word alle Vorkommmen von character
       und gibt das resulierende Wort zurueck
    '''


def replace_char(word, character, new_character):
    '''ersetzt in word jedes Vorkommen von
       character durch new_character
    '''

    
def strip(word):
    '''entfernt allen Space vor und nach dem Wort

    '''
```

````python
def delete_character(word, character):
    '''word: str
       character: str der Laenge 1
       entfernt aus word alle Vorkommmen von character
       und gibt das resulierende Wort zurueck
    '''
    result = ''
    for c in word:
        if c == character:
            continue
        result = result + c

    return result


def replace_char(word, character, new_character):
    '''ersetzt in word jedes Vorkommen von
       character durch new_character
    '''
    result = ''
    for c in word:
        if c == character:
            result = result + new_character
        else:
            result = result + c

    return result


def strip(word):
    '''entfernt allen Space vor und nach dem Wort
       und gibt dann das Wort zurueck
    '''
    SPACE = ' '
    n = len(word)

    for i in range(n):
        if word[i] != SPACE:
            break

    for j in range(1, n+1):
        if word[-j] != SPACE:
            break

    return word[i:n-j+1]


def find(s, sub, start=0, end=None):
    '''gib den ersten Index von Substring sub
       in s[start:end] zurueck oder -1
    '''
    s = s[start:end]
    if s == '' or sub == '':
        return -1              # early return

    n = len(s)
    m = len(sub)

    for i in range(n):
        if s[i:i+m] == sub:
            return start + i   # index, falls gefunden

    return -1                  # falls nicht gefunden
````
