### Aufgaben mit Lösungen
1. Frage den Benutzer solange nach seinem Pin (z.B. '1234'),
   bis der korrekte Pin eingebenen wird.
1. Fordere den Benutzer solange auf, ja oder nein einzugeben, bis er dies tatsächlich tut.
1. Frage den Benutzer maximal 3 Mal nach seinem Pin (z.B. '1234').
   Gib dann 'Login successful!' oder 'Login failed!' aus,
   je nach dem, ob der richtige Pin eingebenen wurde.
   Löse die Aufgabe mit einer For- und einer While-Schleife.

1. Benutze die Funktion `randint` aus dem Modul `random`, um 2 Zufallszahlen $x$ und $y$ zu erzeugen. Implementiere dann eine Funktoin,
  die mit einem While-Loop solange nach dem Produkt dieser Zahlen fragt, bis
die richtige Antwort gegeben wird.  
Löse das Problem mit Ausbrechen aus der While-Schleife, einer Hilfsvariable und einer direkten Bedingung.

```python
answer = ''
while not (answer == 'ja' or answer == 'nein'):
    answer = input('ja/nein:')

print(f'Deine Anwort ist "{answer}".')
```

```python
PIN = '1234'
pin = input('Pin:')

while PIN != pin:
    print('falscher Pin.')
    pin = input('Pin:')

print('Login was successful!')
```

```python
PIN = '1234'

for i in range(3):
    pin = input('Pin:')
    if pin == PIN:
        break
    else:
        print('falscher Pin!')

if pin == PIN:
    print('Login was successful!')
else:
    print('Login failed!')
```

```python
PIN = '1234'
pin = input('Pin:')
is_pin_false = pin != PIN
count = 1

while is_pin_false and count < 3:
    print('falscher Pin!')
    pin = input('Pin:')
    count = count + 1
    is_pin_false = pin != PIN

if is_pin_false:
    print('Login failed!')
else:
    print('Login was successful!')
```

```python
# Frage nach dem Produkt zweier kleiner Zufallszahlen
import random

x = random.randint(1, 10)
y = random.randint(1, 10)
question = f'Wieviel ist {x} mal {y}?'
question
```

```python
def ask_sum_question():
    x = random.randint(1, 10)
    y = random.randint(1, 10)
    question = 'Wieviel ist ' + str(x) + ' Mal ' + str(y) + '?'

    while True:
        answer = int(input(prompt=question))
        if answer == x*y:
            break

    print(f'You got it right: {x}*{y} is {answer}')
```

```python
ask_sum_question()
```

```python
def ask_sum_question():
    x = random.randint(1, 10)
    y = random.randint(1, 10)
    question = 'Wieviel ist ' + str(x) + ' Mal ' + str(y) + '?'

    is_correct = False
    while not is_correct:
        answer = int(input(prompt=question))
        is_correct = answer == x*y

    print(f'You got it right: {x}*{y} is {answer}')
```
