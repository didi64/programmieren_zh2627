### String-Interpolation 
String-Interpolation ist das Einfügen von 
Werten von Variabeln oder Ausdrücken
in ein Stringliteral.


In Python ist die mittels sog. f-Strings direkt machbar.
Stellt man einem Stringliteral ein `f` voran, werden alle in `{}` gefasste Ausdrücke
ausgewertet, ins Strings umgewandelt und eingefügt.

```python
f'Das Quadrat von {x} ist {x**2}'
```

Will man das Symbol `{` oder `}` ausgeben, muss
man da `{{` oder `}}` verwenden.

```python
f'{{x**2}} wird zu {x**2}'
```

Statt Strings zu addieren, benutzt man oft f-Strings:
```python
line_1 = 1. Zeile'
line_2 = 2. Zeile'

# lines = line_1 + '\n' + line_2
lines = f'{line_1}\n{line_2}'
```
