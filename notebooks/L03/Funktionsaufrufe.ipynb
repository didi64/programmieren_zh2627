### Funktionen aufrufen
Damit eine Programmiersprache von praktischem Nutzen ist, muss es möglich sein,
bestehenden und getesteten Programmcode einfach wiederverwenden zu können. 
**Funktionen** ermöglichen genau das. Funktionen ermöglichen es auch, eine komplexe Programmieraufgabe in kleine Teilaufgaben zu zerlegen.
Bevor wir eigene Funktionen definieren, 
betrachten wir, wie wir bestehende Funktionen nutzen/aufrufen können.

Tippen wir einen Funktionsnamen wie `print` in eine Codezelle, wird angezeigt, was für ein Objekt
`print` ist. Diese Information erhalten wir auch mit `type(print)`.

```python
type(print)  # builtin_function_or_method
print        # <function print(*args, sep=' ', end='\n', file=None, flush=False)>
```


**Funktionsaufruf**  
Damit die Funktion etwa macht, muss sie **aufgerufen** werden. 
Der Aufruf einer Funktion `f` sieht so aus:
```python
f(<Argumente>)
```
Dabei sind `<Argumente>` durch konkrete Argumente zu ersetzen. 
`<Argumente>` ist dabei eine möglicherweise leere, komma-separierte Liste von Variabelnamen, 
Ausdrücken oder sog. Schlüssel-Wert Paare `<Variabelname=Ausdruck>`.

Wie eine Funktion aufgerufen werden kan 
(mit wievielen und welchen Argumenten) wird in der Regel angezeigt, 
wenn man Auf den Funktionsnamen klickt und `Shift-Tab` drückt.
Bei `print` erscheint dieser Hilfstext:
```
Signature: print(*args, sep=' ', end='\n')
Docstring:
Prints the values to a stream, or to sys.stdout by default.

sep
  string inserted between values, default a space.
end
  string appended after the last value, default a newline.
```
**Erläuterungen**:
- Die *Signature* zeigt, mit welchen Argumenten die Funktion aufgerufen werden kann.  
  - `*args`: kein, ein oder mehrere Argumente
    ```python
    print()
    print('test')
    print('Zahl:', 5)  # gibt `Zahl:` und 5 aus, getrennt durch ' '
    print(1, 2+2, 10)
    
    ```
  - `sep=' '` zeigt an, dass `sep` ein Default-Argument ist mit Default-Wert `' '`.
    Dieser Wert kann überschrieben werden.
    ```python
    print(1, 2, sep=', .) 
    print(1, 2, sep=', ', end='.')
    ```

Klickt man auf `max` und drückt `Shift-Tab` erscheint der Hilfstext
```
Docstring:
max(iterable, *[, default=obj]) -> value
max(arg1, arg2, *args) -> value

With a single iterable argument, return its biggest item. The
default keyword-only argument specifies an object to return if
the provided iterable is empty.
With two or more positional arguments, return the largest argument.
```

**Erläuterungen**:   
Die Funktion kann man auf 2 Arten verwenden:
1. `max(iterable, *[, default=obj]) -> value`.
   Der Funktion **muss** ein *Iterable* als Argument übergeben werden.  
   Optional kann ein Schlüssel-Wert `default=<obj>` übergeben werden.
   Wie der Hilfstext sagt, wird dieser Wert zurückgegeben, falls das Iterable leer ist (optional, da in   Klammern `[...]`).  
  `-> value` sagt, dass die Funktion einen Wert zurück gibt.

2. `max(arg1, arg2, *args) -> value`.
    Der Funktion **müssen** ein zwei Argument übergeben werden (`arg1` und `arg2`), weitere sind
    sind optional (`*args`).
    Das grösste Argument wird zurückgegeben.
   
**Tipp** Ignoriere alles im Hilfstext, das du nicht verstehst.
**Experimentiere**: Rufe die Funktion auf verschiedene Arten auf, teste was funktioniert.
