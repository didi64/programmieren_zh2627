2. Löse Aufgaben 11-13 aus `L01.Aufgaben_01.ipynb` mit einem For-Loop.

````python
# 2.11 Berechne 1^3+2^3+ ... +n^3
n = 5
total = 0
for n in range(1, n+1):
    total = total + n**3

total
````

````python
# 2.12 bilde einen String der Form numbers '20,26,32,38,44' von
# Zahlen mit konstanter Differenz step von start bis vor stop.
start = 20
step = 6
stop = 45

sep = ','
numbers = ''

for number in range(start, stop, step):
    numbers = numbers + str(number) + sep

numbers[:-1]
````

````python
# 2.13 erstelle String den durch 137 teilbaren Zahlen
# von 500 bis vor 2000
# '548,685,822,959,1096,1233,1370,1507,1644,1781,1918'
start_ = 500
stop = 2000
step = 137

start = start_ - (start_ % step) + step
sep = ','
numbers = ''

for number in range(start, stop, step):
    numbers = numbers + str(number) + sep

numbers[:-1]
````

````python
# Pruefe, dass Zahlen durch 137 teilbar sind
for number in range(start, stop, step):
    # number = number//step)*step + number%step
    msg = (str(number) + ' = ' + str(number // step) + '*'
           + str(step) + ' + ' + str(number % step)
           )
    print(msg)
````

3. Gib die 7-ner Reihe in folgender Form aus. Benutze eine For-Schleife.

```
    1 mal 7 ist 7  
   ...  
   10 mal 7 ist 70  
```   

````python
# 3
for i in range(10):
    n = i + 1
    number = str(n)
    padding = ' '*(2 - len(number))
    line = padding + number + ' mal 7 ist ' + str(n*7)
    print(line)
````

5. Erstelle ein Tuple mit den durch 3 teilbaren Zahlen von 1 bis und mit 99.
Wieviele soche Zahlen gibt es?  
  a) Erstelle das Tuple direkt aus einem `range`-Objekt.  
  b) Benutze eine For-Schleife.
Teste mit `==`, dass beide Tuple die gleichen Elemente enthalten.

````python
# 5a)
numbers_ = tuple(range(3, 100, 3))
numbers_[:3], numbers_[-3:]
````

````python
# 5b)
numbers = ()
for i in range(3, 100, 3):
    numbers = numbers + (i,)

print('Anzahl:', len(numbers))
print('First 3:', numbers[:3])
print('Last 3:', numbers[-3:])
````

6. Spalte ein Wort in zwei Teile `head` und `tail` ohne Slicenotation zu nutzen.  
`head` ist der Teil vor dem $i$-ten Buchstabe, `tail` ist der Rest.  
 

````python
# 6
word = 'foobar'
m = 3

head = ''
tail = ''
n = len(word)

for i in range(m):
    head = head + word[i]

for j in range(i+1, n):
    tail = tail + word[j]

head, tail
````
