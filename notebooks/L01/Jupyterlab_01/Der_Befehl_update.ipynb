### Der Befehl `update <Nummer der Lektion>`
**Files für die aktuelle Lektion herunterladen**  
Mit dem Befehl `update <Nummer der Lektion>`, welcher im Jupyterlab-Terminal einzugeben ist, können die Files einer Lektion heruntergeladen werden. `<Nummer der Lektion>` ist dabei durch eine Zahl zu ersetzen.
Die Nummer der ersten Lektion ist 1. 
- `update 1`  lädt die Files der ersten Lektion herunter
  
Dieser Befehl lädt die neuen Notebooks von einem Git-Repo in den Container und kopiert die Files dann in
den Ordner `work`. Befindet sich bereits ein File mit dem gleichen Namen im Ordner `work`, wird gefragt, ob das File überschrieben werden soll.
Will man bestehende Files überschreiben, ist der Schalter `-y` zu nutzen:
- `update <Nummer der Lektion>` fragt vor dem Überschreiben,
- `update -y <Nummer der Lektion>` überschreibt immer.


Es können auch einzelne Files oder Ordner einer Lektion heruntergeladen werden. Z.B.
- `update L01/Jupyterlab_Intro.ipynb`
- `update L01/Aufgaben`

Manchmal nutzen wir den Befehl auch so:
- `update -e`  lädt Extras herunter
- `update -s`  lädt Scratch-Files herunter (Files, die während einer Lektion entstehen)
