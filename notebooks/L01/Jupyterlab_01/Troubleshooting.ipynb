### Troubleshooting (Kernel shout-downs)
Verhält sich das Notebook merkwürdig, hilft in den meisten Fällen ein
Kernel-Restart (⟳ in der Menuleiste oben).  

Gelegentlich sind auch die Kernels aller Notebooks herunterzufahren, oder
die Language Servers sind herunterzufahren und neu zu starten: 
- Klicke ganz links auf den Knopf 🔘 (oder links unten auf 🔲), dann *Shut all Down* im Feld *Kernels* oder *Language Servers*.  Um die *Language Servers* dann wieder neu zu starten ist der
  **Page-Reload Button des Browsers** zu Klicken!

**Wann ist dies angezeigt**  
- Kernel shut-downs:  
Jedesmal, wenn ein Notebook geöffnet wird, wird ein **neuer** Python-Kernel gestartet, welcher für die Ausführung der Codezellen dieses Notebooks zuständig ist.
Unten links wird die Anzahl aktiver Kernels angezeigt. 
Sind viele Kernels aktive, kann das Jupyterlab langsam werden.  
- Language Server restart:
  Will man z.B. Variabeln umbenennen und das Contextmenu nach Rechtsklick auf die Variable zeigt
  kein *Rename Symbol* an, ist in der Regel das Notebook nicht mit dem Language Server verbunden.
  In diesem Fall sieht man ganz unten neben `<>` den Text *waiting*,
  aber Warten hilf nicht.
