### Unterhaltung mit dem Python-Interpreter

Es gibt kompilierte und interpretierte Sprachen.  

**[Kompilierte Sprachen](https://en.wikipedia.org/wiki/Compiled_language)**:  
Ein Programm ist im wesentlichen eine Folge von Symbolen. Ein Text, der nach bestimmten Regeln geformt ist.
Ein Übersetzungsprogramm, der Kompiler, produziert aus diesem Text eine Folge von Nullen und Einsen, 
ein **Binary** (binäres File), welches Anweisungen enthält, die direkt vom Prozessor/Betriebssystem ausgeführt werden können.

Beispiele: C, C++.

**[Interpretierte Sprachen](https://en.wikipedia.org/wiki/Interpreter_(computing))**:  
Hier wird der Programmcode nicht in ein Binary übersetzt. 
Stattdessen führt der Interpreter (auch ein Programm) Anweisung um Anweisung des Programmcodes aus.  

Beipiele: Python, JavaScript.

**Kompilierte Programme** sind **schneller** als interpretierter Code. 
Das kompilierte Programm ist optimiert auf eine bestimmte Prozessorarchitektur und Betriebssystem.
**Interpretierte Sprachen** sind dafür **plattformunabhängig**, da der Interpreter die Anweisungen des Programms selber ausführt.

### Python-Interpreter in Jupyterlab-Terminal starten

Klicke auf <span style="background-color:blue; color:white">&nbsp;&nbsp;+&nbsp;&nbsp;</span>  und dann unter *Other* auf Terminal. Tippe dann im Terminal den Befehl $\texttt{python}$, um den Interpreter im interaktiven Modus zu starten, dem sog. Read-Evaluate-Print Loop (REPL).
```
studi@jlab $ python
Python 3.13.0 (main, Dec  8 2024, 01:54:23) [GCC 9.4.0] on linux
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```

Nach dem Prompt `>>>` kann man nun Python Code eingeben und ausführen lassen, z.B.
`print('test')`.

- Der Interpreter liest und evaluiert die Eingabe und produziert entsprechenden Output. 
Versteht der Interpreter die Anweisung nicht oder kann er diese nicht ausführen, wird eine Fehlermeldung angezeigt.  
- Mit ↑ und ↓ kann durch die Eingabehistory geblättert werden.
- Der REPL kann mit **ctr-d**  oder dem Befehl **exit** verlassen werden.
