### Integers, Variabelnamen, Strings und Funktionen
**Integers und Operationen mit  Integern**: Ganze Zahlen `-2`, `0`, `42`, ...  
- Negation `-3`, Addition `2 + 2`, Subtraktion `3 - 1`, Multiplikation `2 * 2`, Division `1 / 3` Potenzieren `2**10`, Ganzzahldivision `23 // 5`, module bez. Divisionsrest bestimmen `23 % 5`.  
  Das **Resultat** dieser Operationen ist ein **Integer**.
- Minimum `min(2, 4)`, Maximum `max(-3, 5)`, Betrag: `abs(x)`.
  
Division liefert **immer** ein `float` (Fliesskommazahl), kein Integer!

**Floats und Ops mit Floats** (Fliesskommazahlen): `0.0`, `3.1415926353`  
`round(x)`: rundete die Fliesskommazahl auf den nächsten Integer.

**Variabelnamen** bestehen aus den Zeichen
   > `A..Za..z0..9` und `_` (Gross- und Kleinbuchstaben, Ziffern und Underscore)  
     dürfen aber **nicht  mit einer Ziffer beginnen**.  
     **Keywords** dürfen **nicht** als Namen verwendet werden!

Einer Variable)kann man mit dem Zuweisungsoperator `=` einen Wert zuweisen.

```python
x = 42
```
Tippt man einen Variabelname  in eine Codezelle,
wird dieser Token als Name erkannt.
Der Python Interpreter schaut, ob dieser Variable bereits ein Wert zugewiesen wurde.
Falls **nicht**, wird ein `NameError` ausgelöst.

**Namenskonventionen für Variabeln**:  Der Variabelname sollte nahelegen,
wie die Variable zu verwenden ist.


* `ALL_CAPS` für Konstanten.
* `snake_case` für Funktionen, Variablen und Modulnamen


**Strings und Ops mit Strings**  sind Zeichenkette (Text).
Damit er Interpreter eine Zeichenkette erkennt (und nicht als Name)
muss sie in Anführungszeichen gesetzt werden.
Dabei kann man *single Quotes* `'das ist ein String'` oder *double Quotes* `"das ist auch ein String"` verwenden.  

- Addition `'foo' + 'bar'`. Strings zusammenhängen.
- `len(s)`. Liefert die Länge (Anzahl Zeichen) des Strings `s`.
- Multiplikation `50 * '-'`. Multiplikation mit $0$ oder einer negativen ganzen Zahl ergibt den leeren String `''`.
-Indexing `s[i]`. $i$-tes Zeichen herausgreifen (Das $0$te Zeichen ist das erste Zeichen). Enthält der String das entsprechende Zeichen nicht, wird ein Fehler (`IndexError`) ausgelöst.
- Slicing: `s[2:4]`, `s[:2]`, `s[2:]`, `s[:]`
  Teilstring herausgreifen. Der Index vor dem Doppelpunkt ist das
  erste Zeichen, Der Index nach dem Doppelpunkt ist das erste Zeichen, das **nicht mehr** herausgegriffen wird.

**Funktionen**: Funktionen sind Python-Objekte, die man aufrufen kann.
```python
   print('Hallo Welt!')
   len('Hallo Welt!')
   max(2, 5)
```

Der Aufruf einer Funktion `f` sieht so aus:
```python
f(<Argumente>)
```
Dabei sind `<Argumente>` durch konkrete Argumente zu ersetzen. 
`<Argumente>` ist dabei eine möglicherweise leere, komma-separierte Liste von Variabelnamen.
Eine Funktion gibt immer einen Wert zurück. Dieser Wert kann der spezielle Wert `None`
sein, der das Fehlen eines Wertes anzeigt.

### String-Surgery
Strings sind **immutable** (nicht modifizierbar).
Python erlaubt nicht, ein einzelnes Zeichen eines Strings zu ändern.  
Will man ein Zeichen aus einem String entfernen und in den String einfügen,
muss man den String zerlegen und neu zusammensetzen.

```python
word = 'Teat'
word_1 = word[:2] + 's' + word[3:]
word_2 = word[:1] + word[2:]
word_1, word_2  # Test, Tat
```
***

````python
s = 'Teat'
i = 2  # Index von 'a'
s[i]
````

````python
teil_bis_a = s[:i]
teil_bis_a
````

````python
teil_nach_a = s[i+1:]
teil_nach_a
````

````python
teil_bis_a + 's' + teil_nach_a
````

### Steuerzeichen

Ein String kann auch **Steuerzeichen** wie Zeilenumbrüche enthalten.

Ein Zeilenumbruch wird mit `\n` (Backslash + n) codiert.
Intern wird ein String als Folge von Bytes gespeichert, z.B.\
`HALLO\n` -> `48 41 4C 4C 4F 0A`.\
Das Byte `0A` codiert dabei das Steuerzeichen `\n`.

```python
zeile_1 = 'Erste Zeile.'
zeile_2 = 'Zweite Zeile.'
width_1 = len(zeile_1)
width_2 = len(zeile_2)
width = max(width_1, width_2)
footer = width * '-'

text = (zeile_1 + '\n'    # Ein Ausdruck in Klammern kann 
        + zeile_2 + '\n'  # sich ueber mehrere Zeilen erstrecken!
        + footer
        )

print(text)
```
