### Die Funktion `print`

Viele Variabeln haben bereits einen Wert.

Sog. `builtins`, werden gr&uuml;n angezeigt

In Python ist der Wert einer Variable ist immer ein sog. **Objekt**.


Auch Integers (`42`) und Strings (`'hallo'`) sind Objekte.
Manche Objekte sind `aufrufbar` (callable).
Unter anderen sind Funktionen aufrufbar.

Ein Objekt ist im Python-Sinne "aufrufbar", wenn man runden Klammern `()` an es anhängen kann, 
um Code auszuführen. Unter anderen sind **Funktionen** (wie `print`) aufrufbar.
