6. Weise einer Variable einen String zu, der folgenden obigen Output produziert, wenn er mit `print` ausgegeben wird.
```
##################################################
#                                                #
#                Text inside box.                #
#                                                #
##################################################

```
```python
SHARP = '#'
SPACE = ' '
NEWLINE = '\n'

width = 50
text = 'Text inside box.'
...
```

````python
SHARP = '#'
SPACE = ' '
NEWLINE = '\n'

width = 50
text = 'Text inside box.'

n = len(text)
n_spaces = (width - 2 - len(text)) // 2

top_bottom = width * SHARP
extra_line = SHARP + (width - 2) * SPACE + SHARP
text_line = SHARP + n_spaces * SPACE + text + n_spaces * SPACE + SHARP
paragraph = (top_bottom + NEWLINE + extra_line + NEWLINE
             + text_line + NEWLINE + extra_line + NEWLINE + top_bottom)

print(paragraph)
````
