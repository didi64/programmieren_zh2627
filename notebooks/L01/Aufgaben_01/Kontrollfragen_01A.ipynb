### Kontrollfragen MiniPython II
1. Schreibe Mini-Python Code der folgenden Output produziert, wobei sich die Breite der Box nach
   dem Wert einer Variable `width` richtet.
```
##################################################
#                                                #
#                Text inside box.                #
#                                                #
##################################################

```
```python
SHARP = '#'
SPACE = ' '

width = 50
text = 'Text inside box.'
...
```
2. Weise einer Variable einen String zu, der ebenfalls obigen Output produziert, wenn er mit `print` ausgegeben wird.
```python
NEWLINE = '\n'
...
```

````python
SHARP = '#'
SPACE = ' '

width = 50
text = 'Text inside box.'

n_chars = len(text)
n_spaces = (width - n_chars - 2) // 2  # Anzahl Space vor und nach 'Textbox '

line_1 = width*SHARP
line_2 = SHARP + (width - 2)*SPACE + SHARP
text_line = SHARP + n_spaces*SPACE + text + n_spaces*SPACE + SHARP

print(line_1)
print(line_2)
print(text_line)
print(line_2)
print(line_1)
````

````python
NEWLINE = '\n'
header = line_1 + NEWLINE + line_2
footer = line_2 + NEWLINE + line_1
output = header + NEWLINE + text_line + NEWLINE + footer
print(output)
````
