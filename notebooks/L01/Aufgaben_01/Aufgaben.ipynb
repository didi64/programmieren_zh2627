**Newton Verfahren**

**Beispiel**: $f(x)=x^2 - 2$**:  
Ableitung $f'(x) = 2x$. Wir erhalten
$$
x_{n+1} = 
x_n  - \frac{f(x_n)}{f'(x_n)} = 
x_{n+1} = 
x_n  - \frac{x^2_n - 2}{2x_n} =
x_n - (\frac{x_n}{2} -\frac{1}{x_n}) =
\frac{x_n}{2} + \frac{1}{x_n}.
$$

```python
Can you give we a python regex that
replaces '&auml;' by 'ä', '&ouml;': by'ö' and '&uuml;' by'ü' and the same for the upper cases
```

```python
import re
```

```python
def decode_umlauts(text):
    umlaut_mapping = {
        '&auml;': 'ä', '&ouml;': 'ö', '&uuml;': 'ü',
        '&Auml;': 'Ä', '&Ouml;': 'Ö', '&Uuml;': 'Ü'
    }
    pattern = re.compile('|'.join(re.escape(key) for key in umlaut_mapping.keys()))
    return pattern.sub(lambda match: umlaut_mapping[match.group(0)], text)
```
