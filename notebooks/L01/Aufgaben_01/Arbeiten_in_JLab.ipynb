***
$\sqrt{2}$ mit den **Newton Verfahren** approximieren:  
Die Nullstelle einer Funktion $f(x)$ findet man oft mit folgender Iterationsvorschrift:  
- $x_0 = \text{grobe Schätzung}$,
- $x_{n+1} = x_n  - \frac{f(x_n)}{f'(x_n)}$.

**Beispiel 1**: $f(x)=x^2 - 2$. Die Nullstelle von $f$ ist $\sqrt{2}$.  
Die Ableitung $f'(x) = 2x$. Wir erhalten
$$
x_{n+1} = 
x_n  - \frac{f(x_n)}{f'(x_n)} = 
x_{n+1} = 
x_n  - \frac{x^2_n - 2}{2x_n} =
x_n - (\frac{x_n}{2} -\frac{1}{x_n}) =
\frac{x_n}{2} + \frac{1}{x_n}.
$$

**Beispiel 2**: $f(x)=x^2 - 5$. Die Nullstelle von $f$ ist $\sqrt{5}$. Diesmal erhalten wir
$$
x_{n+1} = \frac{x_n}{2} + \frac{5}{2x_n}.
$$
