### Markdown
Markdown ist eine einfache Auszeichnungssprache.
Überschriften, Hervorhebungen, Listen, Aufz&auml;hlungen, Code, ... lassen sich
einfach in ein Dokument einf&uuml;gen.

* Zeilenumbruch `\` plus neue Zeile oder 2xSpace `  `
* Inline-Code: `` ` `` Text `` ` ``
* Überschriften: `# Überschrift 1` bis `### Überschrift 3`
* Fett: `**fett**`
* Kursiv: `*kursiv*`
* Aufz&auml;hlungen: Mit einem Bindestrich `-` oder Sternchen `*`
* Ungeordnete Liste: Mit -, * oder + einleiten
* Geordnete Liste: Mit 1., 2., 3. etc. einleiten
* Links: `[www.python.org](https://www.python.org)`
* Zitate: `>` Text,
 * Python Code:\
    ```` ```python ````\
     Code\
     ```` ``` ````

**Beispiel** *Python-Code*
```python
x = 2
y = 3
z = x + y
```

### Aufgabe
Doppelklicke in diese und obige Markdownzelle, wirf einen Blick auf den
Markdown-Code und f&uuml;hren dann die Zelle mit `Ctr-Enter` oder `shift-Enter` aus.

Absolviere das Markdown Tutorial: `Help -> Markdown Reference`.
