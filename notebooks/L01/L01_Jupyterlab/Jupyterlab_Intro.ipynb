### Zellen
Ein Notebook besteht aus Zellen.Es gibt **3 Arten** von Zellen. 
Diese ist eine Markdownzelle. Nach deinem Klick in die Zelle 
zeigt der Toolbar diese Festers den Zellentyp an.
Der default Zellentyp ist *Code*. Der Zellentyp kann im Toolbar dieses Fensters ausgewählt werden (klicke auf ⌄ ).

- *Markdown*: Diese Zelle ist eine Markdown-Zelle.  
  Doppelklick auf eine Markdown Zelle zeigt den Markdown-Code an. Der Code lässt sich nun editieren.
  **Ctrl-Enter** führt den Markdown Code aus,und produziert Markdown-Text. Der Fokus bleibt auf der Zelle.
  **Shift-Enter** führt ebenfalls die Zelle aus. Der Fokus **springt** zur **nächsten Zelle**.
- *Code*: Default. Zelleninhalt wird als Python-Code interpretiert.  
          **Ausser** Text nach `#` (Kommentar). \

  Der Language-Server zeigt "schlechte" (nicht [Pep-8](https://peps.python.org/pep-0008/#introduction) konforme) Formatierung (orange unterwellt) und Syntaxfehler (rot unterwellt) an.
  Orange **sollte** weg. Rot **muss** weg! Roter Code löst eine
  Fehlermeldung aus (meistens ein `SyntaxError` oder `NameError`).

  **shift-Enter** bez. **Ctrl-Enter** führt den Code in der Code-Zelle aus.  
  **Ctrl-z** macht Änderungen **in der Zelle**  rückgängig.  

- *Raw*: Zelleninhalt ist einfach Text.  
  Passiert nichts wenn man Code ausführt ist meisten der Zellentyp
  auf Raw gesprungen (wegen eines Keyboardshortcuts).
- **Keyboardshortcuts**: ESC-r, ESC-y, ESC-m wechseln den Zelltyp auf Raw, Code bez. Markdown.

```python
# Dies ist eine Code Zelle
1 + 2  # berechnet den Wert des Ausdrucks 1 + 2 und gibt in aus
```

```python
'hello world!'  # gibt den String 'hello world!' aus.
```

### Kernel unterbrechen und neu starten, allen Output löschen
- Wird in einer Zelle Code ausgeführt kann nicht gleichzeitig in einer anderen Zelle
  Code ausgeführt werden.  
  [*] rechts neben der Zelle zeigt an, dass Code ausgeführt wird. 
- Die Ausführung von Code stoppen:  
  Klicke oben im diesem Fenster auf `Interrupt Kernel` (&#9209;).    
  Falls das nicht hilft, klicke auf `Restart Kernel` (&#10227;).  
  **Achtung**: Bei einem Kernel restart werden alle Variabeln gelöscht! Gleicher Effekt, wie wenn man das Notebook schliesst und wieder öffnet.
- Verhält sich das Notebook merkwürdig, versuche `Restart Kernel` (&#10227;) oder  
  klicke `Kernel ->  Restart Kernel and Clear All Outputs...`
- Rechtsklick in eine Zelle öffnet ein Kontextmenu, welches erlaubt,
  allen Output zu löschen (Clear Outputs of all Cells/ Clear Cell Outputs).
 

```python
# Endlosschleife, Ausführung muss mit ⏹ unterbrochen werden!
x = 0

while x >= 0:
    x = x + 1

x
```

***
- Wird der Kernel mit ⏹ unterbrochen, behalten die Variabeln ihre Werte
- Wird der Kernel mit ⟳ neu gestartet, sind **alle Variabeln undefiniert**.
