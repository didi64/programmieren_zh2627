### Divide and Conquor

````python
# guess zu klein: lower_bound = max(lower_bound, guess+1)
#                 upper_bound = upperbound
# guess zu gross: lower_bound = bleibt
#                 upper_bound = min(upperbound, guess-1)
````

````python
# lower_bound <= secret <= upper_bound
lower_bound = 0
upper_bound = 100
secret = 42  # zu erratende Geheimzahl
````

````python
guess = 24

feedback = (guess == secret and f'Gratuliere! {guess} war richtig!'
            or guess < secret and f'{guess} ist zu klein'
            or f'{guess} ist zu gross'
            )

lower_bound = guess < secret and max(lower_bound, guess+1) or lower_bound
upper_bound = guess > secret and min(upper_bound, guess-1) or upper_bound

info = f'{lower_bound} < secret < {upper_bound}'
hint = f'try next: {(lower_bound + upper_bound) // 2}'
feedback, info, hint
````
