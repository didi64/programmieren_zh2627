### Erste Programme
Der Ablauf eines "interessanten" Programms wird
mittels **Schleifen** und/oder **bedingten Anweisungen** 
gesteuert. 
Da wir diese Sprachelemente noch nicht eingeführt haben,
emulieren wir eine Schleife durch mehrfaches Ausführen einer Codezelle von
Hand.

Alle nachstehenden Programme haben die gleiche Form.
1. **Initialisierung**: In einer ersten Zelle werden bestimmte Variabeln auf Startwerte gesetzt.
  Diese Zelle ist **einmal** auszuführen
2. **Schleife**: Die zweite Zelle ist die Schleife. Dies Zelle ist mehrfach auszuführen.
  Die Kommentarzeile gibt an, wie oft.
  Oft wir in der Schleife eine Variable inkrementiert oder dekrementiert.
3. **Aufbereitung und Ausgabe** des Resultates.

***

**Cell-Magic**  
In einem ersten Durchlauf sind alle Zellen tatsächlich von Hand mehrfach auszuführen.
Beim 2. Durchlauf wollen wir Cell-Magic nutzen. Beginnt die erste Zeile einer Code-Zelle mit
  > `%%<magic-command>`

  wird der Befehl `<magic-command>` auf die Zelle angewendet.
  
  
  ```python
  %%run_cell 5 # führt die Zelle 5 mal aus 
               # (statt 5 kann man auch eine (definierte) Variable verwenden)
  ...
  ```
Jupyterlab stellt eine Anzahl solcher Cell-Magics zur Verfügung, und es lassen sich
eigene Cell-Magics programmieren (wie z.B. `run_cell`). 
Um `run_cell` nutzen zu können, muss diese Funktion zuerst importiert werden.
Dazu ist nachstehende Code-Zeile auszukommentieren und auszuführen.  
Auch die Cell-Magic ist auszukommentieren (`#%%run_cell n` -> `%%run_cell n`).

````python
# from magic import run_cell
````

***
$n$ Fakultät berechnen ($n\cdot(n-1)\cdots 2\cdot 1$) 
***

````python
n = 5
result = 1
````

````python
#%%run_cell n - 1 # times
result = result * n
n = n - 1
````

````python
result, n
````

***
Summe der ersten $n$ Zahlen $(n + (n-1)+\ldots+2 + 1$).
***

````python
n = 5
total = 0
````

````python
#%%run_cell n # times
total = total + n
n = n - 1
````

````python
total, n
````
