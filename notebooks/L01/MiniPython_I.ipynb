In Python gibt es viele verschiedene Arten (Typen) von Werten.
Nachfolgend betrachten wir **Integers** (ganze Zahlen), 
**Floats** (Fliesskommazahlen)
**Strings** (Text/Zeichenketten) und die Wahrheitswerte (booleans)
`True` und `False`.  

**Integers**  
Integer sind ganze Zahlen wie -100, 0, 1, ... In Python kann ein Integer **bis zu 4'300 Stellen** haben.


**Strings**  
Ein String ist Text, bez. eine Zeichenkette.
Damit eine Zeichenkette als String aufgefasst wird, muss sie in Anführungszeichen gesetzt werden.
Dabei kann man *single Quotes* `'das ist ein String'` oder *double Quotes* `"das ist auch ein String"` verwenden.  

**Steuerzeichen**

Steht innerhalt der Anführungszeichen `\n` (ein  Backslash gefolgt von `n`), so wird diese Kombination bei der Ausgabe mit `print` als Zeilenumbruch interpretiert (siehe unten bei Ausgabe).

  
Die **Wahrheitswerte/Booleans** `True` und `False`  
Die Werte `True` und `False` stehen für wahr und falsch.  
`True` und `False` verhalten sich genau wie `1` und `0`.


**Operationen mit  Integern** 
- Addition (`2 + 2`), Subtraktion (`3 - 1`), Multiplikation (`2 * 2`), Division (`1 / 3)`  
  Potenzieren (`2**10`), Ganzzahldivision (`23 // 5`), Divisionsrest bestimmen (`23 % 5`).  
  Das **Resultat** dieser Operationen ist jeweils wieder ein **Integer**.
- Test auf Gleichheit (`2 == x`), Ungleichheit (`2 != x`), grösser (`x < y`),  grösser gleich (`x <= y`), und entsprechend kleiner und kleiner gleich.  
  Test ob ein Wert zwischen zwei anderen liegt: `x < y < z`.
  Test ob mehrere Werte gleich sind: `x == y == z`.
  Das Resultat ist jeweils einer der Wahrheitswerte `True` oder `False`.
- Minimum `min(2, 4)` und Maximum `max(-3, 5)`.
- Division (`4 / 2`) liefert **immer** ein `float` (Fliesskommazahl), kein Integer!
- 
**Operationen mit  Strings**
- Konkatenation (`'foo' + 'bar'`). Das Resultat is wieder ein String.
- Mit `len` kann man die Länge eines Strings bestimmen. Das Resultat ist ein Integer.  
- Test auf Gleichheit und Ungleichheit und grösser/kleiner (gleich).  
  Der "kleinere" String kommt alphabetisch vor dem "grösseren" String.

**Operationen mit  Strings und Integers**  
- String vervielfachen (`'-' * 80`). Multiplikation eines Strings mit einer negativen Zahl ergibt den leeren String `''`.
-  Zeichen aus String `s` herausgreifen (`s[0]`, `s[1]`, `s[2]`).  
   Das $0$te Zeichen ist das erste Zeichen. Enthält der String das entsprechende Zeichen nicht, wird ein Fehler (`IndexError`) ausgelöst.

**Operationen mit Wahrheitswerten**  
- Einen Wahrheitswert kann man mit `not` negieren. Zwei   Wahrheitswert kann man mit `and` oder `or` verknüpfen.
- Mit `is` bez. `is not` kann man testen, ob zwei Wahrheitswerte gleich oder verschieden sind.
