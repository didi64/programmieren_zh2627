### Tokens: Die syntaktischen Bausteine einer Programmiersprache 
Bevor der
Python-Interpreter den Code ausführen kann, analysiert er den Code.
In einem ersten Schritt zerlegt er den Code in sog.  **Tokens**, kleinste sinnvollen Bausteine eines Programms.

Die Programmiersprache Python unterteilt Quellcode in fünf Haupttypen von Tokens. 

1. **Keywords** (Schlüsselwörter): Reservierte Wörter mit fester Bedeutung (z. B.
   `class`, `def`, `else`, `for`, `from`, `import`, `return`, `while`, ...).
4. **Identifiers** (Namen): Vom Programmierer oder von Python selbst vergebene Namen für (normale) Variablen, Funktionen, Klassen oder Module (z. B. `x`, `print`, `user_name`, `is_even`). Gültige (Variabel)namen bestehen aus den Zeichen
   > `A..Za..z0..9` und `_` (Gross- und Kleinbuchstaben, Ziffern und Underscore)  
     dürfen aber **nicht  mit einer Ziffer beginnen**.  
     **Keywords** dürfen **nicht** als Namen verwendet werden!

   **Bemerkung**: Funktionsnamen (Klassennamen, ...) sind Variabeln,
   die Funktionen (Klassen, ...) speichern. Etwas ungenau spricht man in der Regel aber nicht von
   Funktionsnamen, sondern von Funktionen.
6. **Literals** (Literale): Feste Werte im Code:
	* Numerische Literale (Ganze Zahlen (`42`), Gleitkommazahlen (`3.14159`))
	* String Literale (Strings `'Hello'` oder `"Test"`)
	* Boolesche Literale (`True`, `False`)
	* Der spezielle Wert `None`
7. **Operators** (Operatoren): Symbole für Berechnungen oder Vergleiche:
	* Arithmetische Operatoren (`+`, `-`, `*`, `/`, `**` (hoch), `%` (module), `//` (Ganzzahldivision))
	* Vergleichs- und Zuweisungsoperatoren (`=`, `==`, `!=`, `<`, `>`)
	* Logische Operatoren (`and`, `or`, `not`)
8. **Punctuators / Delimiters** (Satz- und Trennzeichen): Symbole zur Strukturierung des Codes:
	* Klammern und Blöcke (`(`, `)`, `[`, `]`, `{`, `}`)
    * Zeilenumbrüche (Newlines)
	* Doppelpunkte und Kommas ( `:`, `,` )
	* Einrückungen (INDENT, DEDENT) und Zeilenumbrüche, welche in Python den Programm-Block definieren

**Bemerkung**: In  Python sind **Einrückungen** Teil der Sprache! Andere Sprachen
nutzen z.B. Klammern `{` und `}` um Code zu einem Block zusammenzufassen. 
In Python definieren **Einrückungen** die Code-Blöcke!

```python
for name in ('Alice', 'Bob', 'Carl'):
    print(name)  # Schleifenbody, muss eingerueckt sein!
```
