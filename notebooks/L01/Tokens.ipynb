### Tokens: Die syntaktischen Atome der Programmiersprache 


Die Programmiersprache Python unterteilt Quellcode in fünf Haupttypen von Tokens 
(kleinste sinnvollen Bausteine eines Programms):

1. **Keywords** (Schlüsselwörter): Reservierte Wörter mit fester Bedeutung (z. B. `if`, `else`, `for`, `while`, `def`, `class`, `import`, `return`), die nicht als Namen für Variablen verwendet werden dürfen.
2. **Identifiers** (Namen): Vom Programmierer oder von Python selbst vergebene Namen für Variablen, Funktionen, Klassen oder Module (z. B. `print`, `user_name`, `is_even`). G&uuml;ltige Variabelnamen bestehen aus den Zeichen
   > `A..Za..z0..9` und `_` (Gross- und Kleinbuchstaben, Ziffern und Underscore)  
     d&uuml;rfen aber **nicht  mit einer Ziffer beginnen**. 
3. **Literals** (Literale): Feste Werte im Code:
	* Numerische Literale (Ganze Zahlen (`42`), Gleitkommazahlen (`3.14159`))
	* Zeichenketten-Literale (Strings `'Hello'` oder `"Test"`)
	* Boolesche Literale (`True`, `False`)
	* Der spezielle Wert `None`
4. **Operators** (Operatoren): Symbole für Berechnungen oder Vergleiche:
	* Arithmetische Operatoren (`+`, `-`, `*`, `/`, `**` (hoch), `%` (module), `//` (Ganzzahldivision))
	* Vergleichs- und Zuweisungsoperatoren (`=`, `==`, `!=`, `<`, `>`)
	* Logische Operatoren (`and`, `or`, `not`)
5. **Punctuators / Delimiters** (Satz- und Trennzeichen): Symbole zur Strukturierung des Codes:
	* Klammern und Blöcke (`(`, `)`, `[`, `]`, `{`, `}`)
	* Doppelpunkte, Kommas und Semikolons ( `:`, `,`, `;` )
	* Einrückungen (INDENT, DEDENT) und Zeilenumbrüche, welche in Python den Programm-Block definieren

**Bemerkung**: In  Python sind **Einrückungen** Teil der Sprache!
