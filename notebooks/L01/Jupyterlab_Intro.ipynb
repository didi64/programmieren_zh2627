### Zellen
Ein Notebook besteht aus Zellen.Es gibt 3 Arten von Zellen.
Der default Zellentyp ist *Code*. Der Zellentyp kann im Toolbar dieses Fensters ausgewählt werden (klicke auf ⌄ in der Menuleiste dieses Fensters).
- *Code*: Zelleninhalt wird als Python-Code interpretiert.

  Der Language-Server zeigt "schlechte" (nicht [Pep-8](https://peps.python.org/pep-0008/#introduction) konforme)Formatierung (orange unterwellt) und Syntaxfehler (rot unterwellt) an.
  Orange **sollte** weg. Rot **muss** weg! Roter Code l&ouml;st einen `SyntaxError` aus.

  **shift-Enter** führt den Code in der Code-Zelle aus und springt zur nächsten Zelle.  
  **Ctrl-Enter** führt den Code in der Code-Zelle aus und behält den Fokus auf dieser Zelle.
  **Ctrl-z** macht Änderungen **in der Zelle**  rückgängig.  

- *Markdown*: Diese Zelle ist eine Markdown-Zelle.  
  Doppelklick auf eine Markdown Zelle zeigt den Markdown-Code an. Der Code lässt sich nun editieren. **Ctrl-Enter** produziert aus Markdown-Code dann wieder Markdown-Text.
- *Raw*: Zelleninhalt ist einfach Text.


```python
# Dies ist eine Code Zelle (Text nach # ist Kommentar)
1 + 2  # zeigt den Wert des Ausdrucks '1 + 2' an
```

```python
print('Hello world!')  # gibt den String Hello world!' aus.
```

### Fehlermeldungen
Wird eine Codezelle ausgef&uuml;hrt und von Interpreter nicht verstanden, wird ein
Fehler ausgel&ouml;st und eine ev. lange Fehlermeldung auf rotem Grund angezeigt.  

Die **wichtigste** Information steht ganz am Schluss:
- In welcher Zelle und in welcher Zeile ist der Fehler.
  Die Zeilennummer(n) und die Zeile(n) in der Umgebung des Fehlers werden angezeigt.
- Der Name des Fehlers steht ganz unten.\
  Einer der h&auml;ufigsten Fehler ist ein `NameError`.
  Dieser Fehler wird erzeugt, wenn versucht wird, eine **undefinierte** Variable auszuwerten.

### Kernel unterbrechen und neu starten, allen Output löschen
- Wird in einer Zelle Code ausgeführt kann nicht gleichzeitig in einer anderen Zelle
  Code ausgeführt werden.  
  [*] rechts neben der Zelle zeigt an, dass Code ausgeführt wird. 
- Die Ausführung von Code stoppen:  
  Klicke oben im diesem Fenster auf `Interrupt Kernel` (&#9209;).    
  Falls das nicht hilft, klicke auf `Restart Kernel` (&#10227;).  
  **Achtung**: Bei einem Kernel restart werden alle Variabeln gelöscht! Gleicher Effekt, wie wenn man das Notebook schliesst und wieder öffnet.
- Verhält sich das Notebook merkwürdig, versuche `Restart Kernel` (&#10227;) oder  
  klicke `Kernel ->  Restart Kernel and Clear All Outputs...`
- Rechtsklick in eine Zelle öffnet ein Kontextmenu, welches erlaubt,
  allen Output zu löschen (Clear Outputs of all Cells/ Clear Cell Outputs).
